# Backtracking

### CSP
Un constraint satisfaction problem es una tripleta (X, D, C) tal que:
- X: {x_1, ..., x_n} es un conjunto de variables.
- D: {D_1, ..., D_n} es un conjunti de dominios respectivos.
- C: {C_1, ..., C_n} es un conjunto de restricciones.
donde cada restricción involucra un subconjunto de variables de X. 
Una solución es una asignación de variables en sus dominios tal que se satisfacen todas las restricciones.
\* No necesariamente las variables son del mismo dominio.
\* Una trstricción C_j puede involucrar 1 o más vars de X.

\* ej. Sudoku


### Backtracking
Es una estrategia de búsqueda sistemática que consiste en construir soluciones de forma incremental, probando diferentes configuraciones y retrocediendo cuando se detecta que una solución no es válida. En otras palabras, cada vez que la solución incumple una restricción, nos devolvemos a probar otra configuración del último elemento.
Es la forma central de resolver CSPs.

1. Realizar una asignación de la variable x_k cuando ya se han asignado x_1, ..., x_k-1.
2. Se verifica si la asignación parcial x_1, ..., x_k-1, x_k puede terminar en una solución al problema.
3. Si no es así, nos retractamos y deshacemos la asignación de x_k.

\* Backtracking es igual o más rápido que la fuerza bruta.

\* Queremos determinar si un problema tine solución.

\* Deben ser dominios finitos, variables discretas.

isSolvable(X, D, R):
    if X is empty: return true
    x <- alguna variable de X
    for v in D_x:
        if x = v no rompe R: <!-- Se recomienda hacer una función para verificar restricciones -->
            x <- v
            if isSolvable(X - {x}, D, R):
                return true
            x <- empty
    return false


\* La complejidad es O(K^n), donde K es el tamaño máximo de los dominios. Si bien no podemos determinar si tiene mejor o peor caso, en la práctica es mejor que fuerza bruta.

````python
from random import random

# Determina si el problema tiene solución
def isSolvable(X, D, R):
    if not X:
        return True
    x = random.choice(X) # Elegimos alguna variable de X
    for v in D[x]:
        # Probamos si la asignación x = v cumple las restricciones
        if x == v and check(v, R):
            x = v
            # Verificamos si la asignación parcial puede llevar a una solución
            if isSolvable(X - {x}, D, R):
                return True
            # Si la asignación no lleva a solución, 
            # deshacemos la asignación y probamos con otro valor
            x = None
    # Si probamos todas las combinaciones y no hay solución, indicamos que el 
    # problema no tiene solución
    return False


# Se puede separar la verificación de restricciones de la búsqueda
def check(v, R):
    if R(v) == True:
        return True
    return False
````

Se puede pensar backtracking como la búsqueda en un árbol, donde cada nivel representa una variable y cada rama una asignación posible. Si se llega a un nodo que viola una restricción, se poda esa rama y se retrocede al nodo padre para probar otra asignación.

#### Extensiones de backtracking

##### Encontrar todas las soluciones
Dentro del isSolvable recursivo, en vez de retornar true cuando se encuentra una solución, se almacena la solución y se continúa buscando. Se agregan restricciones en la verificación.

##### Mejoras de desempeño
- Podas: Se pueden podar ramas del árbol de búsqueda que no pueden conducir a una solución válida. Con cada asignación, se agregan restricciones adicionales, recortando el grafo implícito.

- Propagación: Con cada asignación, se pueden actualizar los dominios de las variables restantes. Cuando se asigna una variable (x = v) se propagan los dominios a otras variables (se avisa al resto del problema). Si se determina una asignación como inviable, hay que volver a marcar el dominio como disponible en el backtrack.

- Heurísticas: Se agrega en el código alguna verificación adicional con conocimiento experto, externo a backtracking. Las heurísticas son estrategias para catalogar variables y valores según que tan buenos son. Si la heurísticas es buena, se reduce el espacio de búsqueda. Si no es tan buena, puede aumentar el tiempo de búsqueda.

### Algoritmos codiciosos

Es un algoritmo de optimización que cuando resulta, resuelve el problema de forma lineal, según la cantidad de variables que hayan.
Busca moverse desde un óptimo local a otro, buscando el óptimo global.

Hay 5 tipos de problemas computacionales:
- Problemas de desición (si, no)
- Problemas de búsqueda (secuencia de estados hasta alcanzar un objetivo)
- Problemas de conteo (número de soluciones diferentes)
- Problemas de optimización (mejor solución de acuerdo a alguna métrica)
- Problemas de función (solución explícita)


Los algoritmos codiciosos plantean una estrategia algorítimica basada en el paradigma de subconjuntos
1. Subconjunto S = {s_1, ..., s_n} con _n_ inputs.
2. Queremos un subconjunto S' ⊆ S que satisfaga restricciones.
3. Queremos solución factible que maximice o minimice una función objetivo.
Un subconjunto S' que cumple con las restricciones se llama factible. Una solución que maximiza/minimiza se llama óptima.

Un algoritmo codicioso depende que la estrategia para encontrar la solución es buena.

Los algoritmos codiciosos trabajan en etapas.
- Consideran un input a la vez.
- Una vez se decide sobre un input, la desición es final (no se arrepiente).
- Ninguna desición posterior cambia la actual.

Se debe considerar los input en cierto orden:
- Se usa un procedimiento de selección.
- Si la inclusión del próximo input produce solución infactible, se descarta.
- Si es factible, se incluye en la solución.

El procedimiento de selección se basa en una medida de optimización o estrategia codiciosa.
- Se selecciona un input de forma localmente óptima.
- Esperamos que esa selección nos lleve a una solución globalmente óptima.

\* Un algoritmo codicioso no siempre encuentra la solución óptima.
\* Varias estrategias codiciosas pueden ser plausibles.
\* La mayoría produce soluciones subóptimas.

### Programación dinámica

Es una técnica algorítmica para resolver problemas de optimización dividiéndolos en subproblemas más pequeños y resolviendo cada uno de esos subproblemas solo una vez, almacenando sus soluciones para evitar cálculos redundantes. Es especialmente útil cuando el problema puede descomponerse en subproblemas superpuestos y tiene una estructura óptima.

Estrategia de optimización basado en la existencia de subproblemas que permiten resolver el problema original, los subproblemas se solapan (comparten sub-subproblemas).

\* No es dividir para conquistar, ahí los subproblemas eran disjuntos.

\* La clave está en **recordar** las soluciones a los subproblemas.

_\* El problema define la estrategia, no al reves._

Se define la estrategia de programación dinámica:
- Número de problemas polinimial (idealmente).
- La solución al problema original puede calcularse a partir de subsoluciones.
- Hay un orden _natural_ de los subproblemas (del más pequeño al más grande) y una recurrencia _sencilla_.
- Recordamos las soluciones a subproblemas.


_\* Ver de los ejemplos de clases las condidiciones de término de recursivo e iterativo._

_\* En programación dinámica aprovechar las soluciones ya calculadas y que se solapan, guardar las variables intermedias._

_\* Si tenemos problemas para definir las restricciones, tratar de hacerlas como función._

### Técnicas
- Dividir para conquistar: el problema debe ser divisible, y los subproblemas independientes. Se puede dividir en N subproblemas, no necesariamente tienen que ser 2 iguales. 

- Backtracking: Encuentra **todas** las soluciones factibles. Sirve para cualquier CSP. Es caro y lento, pero es mucho más flexible para modelar problemas.
  - Requiere computar todas las soluciones factibles (puede ser útil en ciertas ocasiones)
  - Muy costoso en tiempo.

- Algoritmos codiciosos: Problema de optimización (CSP), se suele dar la estrategia óptima, o una estrategia para demostrar que no es óptima. El problema se debe poder descomponer, los subproblemas se solapan.
  - Solo si existe estrategia probada.
  - MUY eficientes en tiempo y memoria.

- Programación dinámica: El problema puede descomponerse en subproblemas superpuestos y tiene una estructura óptima.
  - Sólo si hay subestructura óptima.
  - Puede ser eficiente en tiempo y memoria.

\* La estrategia de solución depende del problema.
\* Si no dan estrategia codiciosa en la prueba, probablemente no es algoritmo codicioso.
\* Los problemas de programación dinámica suelen ser problemas de optimización.
