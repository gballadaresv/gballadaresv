# Árboles Binarios de Búsqueda

Un ABB es una estructura que almacena pares (llave, valor) asociándolos mediante punteros según estructura recursiva.

- El nodo puede tener hasta 2 ABBs asociados mediante punteros:
    * Hijo izquierdo.
    * Hijo derecho.

- Cada nodo x tiene a lo más 1 padre x.p
- El nodo sin padre se conoce como raíz
- Cada nodo x tiene hasta 2 punteros que apuntan a sub-árboles:
    * x.left es puntero a hijo izquierdo
    * x.right es puntero al hijo derecho
    * x.p es puntero al padre (si tiene)
- Un nodo sin hijos se conoce como hoja.

##### Eliminación
- Si el nodo es hoja, simplemente se elimina.

- Si queremos eliminar la raíz, lo reemplazamos con el menor de los mayores, o el mayor de los menores.

```python
# A es el puntero al nodo (primer llamado es raíz)
# p es el puntero al padre del nodo (primer llamado es None)
# Dividir para conquistar
def Search(A, p, k):
    # Retorna el puntero al nodo que contiene la clave k, o None (Null) si no existe
    # Como A es nodo, en estricto rigor retorna el puntero a un sub-árbol
    if (not A) or (A.key == k):
        return (A, p)
    if k < A.key:
        return Search(A.left, A, k)
    return Search(A.right, A, k)


def Insert(A, k, v):
    (B, p) = Search(A, None, k)
    if not B:
        B = None # nodo vacío
        B.key = k
        # Conectar B al padre p en la posición adecuada
    B.value = v


# Para eliminar la raíz, hay que buscar el mayor de los menores, o el menor de los mayores
def min(A):
    if not A.left:
        return A
    return min(A.left)

def max(A):
    if not A.right:
        return A
    return max(A.right)

def delete(A, k):
    (D, p) = Search(A, None, k)
    if not D:
        # Si de es hoja
        if not D.left and not D.right:
            D = None
            # eliminar referencia en p
        # Si tiene sólo 1 hijo
        elif D.left and not D.right:
            D = D.left
        elif not D.left and D.right:
            D = D.right
            # eliminar referencia en p
        # si tiene 2 hijos
        else:
            # se reemplaza D por R
            R = min(D.right) # o max(D.left)
            t = R.right
            D.key = R.key
            D.value = R.value
            R = t
```

### AVL

Un árbol binario de búsqueda está AVL-balanceado si:
1. Las alturas de sus hijos difieren a lo más en 1 entre sí.
2. Cada hijo está AVL-balanceado.

Definimos un atributo de balance:
- Si el izq es más alto: x.balance = -1
- Si los hijos tienen la misma altura: x.balance = 0
- Si el der es más alto: x.balance = 1

##### Rotaciones
Cuando insertemos, recorremos hacia arriba del nodo insertado, mientras revisamos la comparación de alturas de los hijos en el camino. 
En la rotación, X es el primer nodo desbalanceado que encontramos cuando subimos, e Y es el hijo de X en la ruta de inserción. Z es el hijo de Y en la ruta de la inserción (sólo si hay un cambio de dirección en la ruta).

*La eliminación es compleja, a veces se prefiere marcar el nodo como eliminarlo en vez de aliminarlo directamente (se puede si se necesita compactar y optimizar espacio).


### Árboles 2-3
Tiene 2 tipos de nodos:
- 2-nodos: tendrán una llave, y si no son hojas tendrán 2 hijos.
- 3-nodos: tendrán dos llaves distintas y ordenadas, si no son hojas tienen 3 hijos. 

Los 3-nodos tienen un subárbol izquierdo, un subárbol derecho, y un subárbol intermedio que tiene valores entre izq y der.

Permite tener las hojas a la misma profundidad, por lo  que garantiza profundidad O(log(n))

##### Árbol de búsqueda 2-3
El nodo puede tener 0, 2 o 3 hijos según si es hoja, 2-nodo o 3-nodo (no puede tener 1 hijo).

Para insertar llaves, debe ser en las hojas.

Si la hoja era un 2-nodo, queda como 3-nodo.

Si la hoja era 3-nodo, se inserta la llave y se separa en dos nodos 2-nodo.

Los árboles crecen de altura sólo cuando se llena la raíz. Entonces se hace split (se sube el nodo central, y las llaves izq y der quedan como hijos).

También tiene operaciones en O(log(n))

# Rojo-negro

Es un ABB que cumple:
- Cada nodo es rojo o negro.
- La raíz del árbol es negra.
- Si un nodo es rojo, sus hijos deben ser negros.
- La cantidad de nodos negros camino a cada hoja desde un nodo cualquiera debe ser la misma.
- Las hojas vacías se consideran nodos negros.

*Un nodo se inserta como rojo. Si queda como hijo de un rojo, habrá que hacer una rotación (y ajustar colores).

# B+
Un parbol B+ de orden _d_ es un ABB que almacenan pares (llave, valor) tal que:
- Los nodos internos sólo guardan llaves .
- La raíz tiene entre 1 y 2d hijos.
- Los nodos internos tienen entre d y 2d hijos.
- Las hojas están a la misma altura y guardan a lo más 2d pares (llave, valor) ordenados por llave.
- Las hojas están conectadas formando una lista doblemente ligada.
\* Siempre está balanceado.
\* Mantiene la eficiencia de búsqueda, inserción y eliminación O(log_2d(#datos)).
\* Procedimientos eficientes de insertar/eliminar datos.
\* Todos los nodos tienen un uso mínimo del 50% (excepto root).

Los nodos que sólo permiten navegar se llaman directorios, los que tienen los pares se llaman datos.

\* Un índice es un método de acceso que optimiza el acceso a los datos para una consulta o conjunto de consultas en particular.
