# Tablas de Hash

##### Diccionario
Es una estructura de datos que:
- Asocia un valor a una llave.
- Obtiene el valor asociado a una llave.
- Actualiza el valor asociado a una llave.
- Elimina de la estructura una asociación llave-valor (a veces).
\* CRUD

\* Los ABB soportan operaciones de diccionario, pero además tienen propiedad de orden.

##### Funciones de Hash
Dado un espacio de llaves  K y un natural m > 0, una función de hash se define como:
    h: K -> {0, ..., m-1}
Dado k pertencetiente a k, h(k) es el valor de hash.

- Permite mapear un espacio de llaves a otro más pequeño.
- Una función de hash no necesariamente es inyectiva.
- Si m < |K|, no es inyectiva.
\* m << |K|

##### Tablas de hash
Una tabla de hash A es una EDD que asocia valores a llaves indexadas usando una función de hash. A es de tamaño m.

\* Una función de hash típica es h(k) = k mod m (hashing modular, se recomienda que m sea primo)

##### Inserción
Si queremos insertar un valor cuyo valor de hash ya está ocupado, podemos insertar el valor de hash, manteniendo un puntero al valor que estaba antes.

Otra forma es con direccionamiento abierto, si mi bucket está ocupado, buscamos hasta que entontremos uno vacío (puede ser sondeo lineal, cuadrado, etc.). Puede producir nuevas colisiones. Una ventaja es que no requiere memoria adicional, desventaja es que no se pueden eliminar datos. Si queremos buscar un dato que no está, basta encontrar una celda vacía.

\* Rehashing != doble hashing


##### Factor de caga
El factor de carga se define como \lambda = n / m
m es el tamaño de la tabla A, y n los valores almacenados.
Mide qué tan llena está la tabla:
- Encadenamiento: \lambda ~1
- Direccionamiento abierto: \lambda > 0.5 resulta en inserciones y búsqueda muy lentas.

##### Rehashing
Si el factor de carga es muy grande, las operaciones se vuelven costosas.
Rehashing es crear una tabla más grande, aproximadamente el doble de grande (primo más cercano a 2m). Requiere definir una nueva función de hash.

\* El 90% del tiempo una tabla de hash es buena idea.
