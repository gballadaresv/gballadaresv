### Selection sort
Selecciona ordenadamante, va y lo busca.
Primero busca y luego inserta.

1. Definir secuencia B, inicialmente vacía.
2. Buscar menor dato x en A.
3. Sacar x de A e insertarlo al final de B.
3. Si quedan valores en A, volver al 2.
4. return B

*No hay mejor ni peor caso, siempre hace lo mismo
*Es O(1) en memoria (in place)

```python
def selectionsort(A: list) -> list:
    B = []
    while A:
        x = min(A)
        B.append(x)
        A.remove(x)
    return B

def SelectionSort(A: list, n: int) -> list:
    # n = len(A)
    # Recorremos el arreglo desde el primer elemento
    for i in range(n):
        min = i
            # Busca el mínimo elemento de la parte no ordenada
        for j in A[i+1:n]:
            if A[j] < A[min]:
                min = j
        # Intercambia el actual por el mínimo encontrado (ordena)
        A[i], A[min] = A[min], A[i]
    return A
```

### Insertion sort
Inserta en el arreglo de salida ordenadamente.
Primero saca lo que salga y luego guarda ordenado.

1. Definir secuencia B vacía.
2. Tomar primer dato x de A.
3. Sacar x de A.
4. Insertar x en B de manera que B quede ordenada.
5. Si quedan datos en A, volver al 2
6. return B



```python
def InsertionSort(A: list, n: int) -> list:
    for i in range(1, n):
        # Todo lo que está antes del i está ordenado
        j = i
        # Para cada elemento, recorro de atrás hacia adelante
        # Lo compara con el anterior (por eso parte de 1 y no del 0)
        # Comparando, si es menor, lo cambio
        # Comienza a comparar de atrás hacia adelante, cambiándolo hasta que esté ordenado
        # Va trayendo desde el final, comparando hasta que esté en su lugar ordenado
        while (j > 0) and (A[j] < A[j-1]):
            A[j], A[j-1] = A[j-1], A[j]
            j -= 1
```

### Merge sort

##### Merge
Parte del supuesto que se tienen 2 listas ya ordenadas. Comparamos el primer elemento de cada sublista, y agegamos a una lista nueva el menor entre ambos (y se saca de la lista correspondiente). Se repite hasta que se acaben los elementos de alguna de las listas, y luego se agregan todos los elementos de la lista restante. Como todo lo anterior ya estaba ordenado, implica que los elementos restantes en la lista original también están ordenados respectos a la lista nueva, con lo que simplemente se agrega todo.

##### Merge Sort
Divide la lista hasta listas de tamaño 1, luego usa Merge para ordenarlas, luego hace Merge con el resultado (listas ordenadas de 2), y le hace Merge a eso, y al resultado le hace Merge... Hasta que pueda hacer Merge de 2 sublistas para obtener la lista final ordenada.

_Dividir para conquistar._

1. Dividir la lista en mitades (y sus sublistas) hasta que queden listas de tamaño 1.
2. Mezclar las minilistas (ordenadas), y mezclar eso hasta tener la lista original ordenada.

```python
def Merge(A:list, B: list) -> list:
    C = []
    # Comparamos el primer elemento de cada lista, y lo agregamos a C
    while A and B:
        if A[0] < B[0]:
            C.append(A.pop(0))
        else:
            C.append(B.pop(0))
    # Dado que agregamos todos los elementos de A o B, agregamos el resto del otro (que ya estaba ordenado)
    C.extend(A or B)
    return C


def MergeSort(A: list) -> list:
    # Si la lista es de tamaño 1, ya está ordenada (caso base)
    if len(A) == 1:
        return A
    # Separar A
    mid = len(A) // 2
    # Aplica MergeSort de forma recursiva sobre cada mitad
    B1 = MergeSort(A[:mid])
    B2 = MergeSort(A[mid:])
    # Habiendo obtenido dos mitades ordenadas, las combinamos y retornamos
    return Merge(B1, B2)
```

##### Búsqueda binaria
Revisa la mitad, si no lo encuentra, lo busca cada mitad recursivamente.

```python
def BinarySearch(A: list, x, i: int, f: int) -> int:
    # x es el elemento a buscar
    # i es la posición inicial, f la final
    if f < i:
        # -1 si no lo encontramos
        return -1
    # reviso la mitad (usa función suelo)
    m = (i + f) // 2
    if A[m] == x:
        # m es la posición del elemento buscado
        return m
    # Si el elemento es menor, busco en la mitad izquierda
    elif x < A[m]:
        return BinarySearch(A, x, i, m - 1)
    # Si el elemento es mayor, busco en la mitad derecha
    else:
        return BinarySearch(A, x, m + 1, f)
```

##### Partition
*Para calcular la mediana con n par, vamos a elegir uno de los valores (no calcular el promedio de ambas).

Primero obteneros un elemento (aleatoriamente) de la secuencia llamado pivote. Revisamos si el pivote es la mediana, sino revisamos si la mediana es menor o mayor al pivote para cada elemento, luego vemos cuál categoría tiene más valores, lo que indica hacia dónde está la mediana. Y sobre esa sección repetimos hasta encontrar el pivote.

```python
from random import random 

def Partition(A: list, i:int, f: int) -> int:
    # elegimos pivote aleatoriamente dentro de A
    p = random(A[i:f+1])
    m, M = [], []
    M.append(p)
    for x in A:
        # si queremos revisar repetidos hay que poner menor/mayor igual en alguno.
        if x < p:
            m.append(x)
            # en estricto rigor x != p
        else:
            M.append(x)
    # redefinimos la lista como la concatenación de la lista de menores y mayores (se ordena)
    A[i:f+1] = m + M
    return i + len(m)
```

### Quick Sort

Selecciona aleatoriamente un pivote dentro del arreglo, luego se separa la lista en mitades desde la posición del pivote, y se aplcia recursivamente, ordenando cada mitad.

Para demostrar correctitud, partimos de la hipóteis que partition es correcto y deja los pivotes ordenados, con lo que hay que demostrar que todos los elementos fueron pivote.


```python
def QuickSort(A: list, i: int, f: int) -> list:
    if i < f:
        # escoge un pitvote de forma aleatoria
        p = Partition(A, i, f)
        # Dividimos la lista por mitades, y se ordena recursivamente cada mitad
        QuickSort(A, i, p - 1)
        QuickSort(A, p + 1, f)
    return A
```

### Counting Sort

Se comienza creando una lista C vacía de largo según el rango de elementos que hayan en A. Luego, cada posición de C contiene la cantidad de ocurrencias en A (frecuencia). Luego se convierte en una lista de frecuencia acumulada (el último debe coincidir con la cantidad de valores a ordenar). 
Finalmente, por cada posición de A (de atrás hacia adelante), vemos el valor que esté en dicha posición (de A), luego vamos a la posición de C de dicho valor (que contiene frecuencia acumulada hasta ese valor), se le resta uno, y eso indica la posición de la lista ordenada en la que irá el elemento de A.

*Necesita saber el rango de datos.

*Es estable

```python
def CountingSort(A: list, k: int) -> list:
    B = [0] * len(A)
    C = []
    # creamos lista C vacía
    for i in range(k):
        C.append(0)
    for j in A:
        C[j] += 1
    # lista C frecuencia acumulada
    for i in range(1, k):
        C[i] += C[i - 1]
    # Agregamos a B el valor-1 de la posición de C que tenga la frecuencia del elemento j de A (según índices de atrás hacia adelante).
    for j in reversed(A):
        B[C[j] - 1] = j
        C[j] -= 1
    return B
```

### Radix Sort

Comienza ordenando el dígito menos significativo (unidades, luego pasa a las decenas, centenas, etc). 

Tiene 2 variantes: LSD (leas significant digit) y MSD (most significant digit).

- LCD: Es útil si todos los strings son del mismo largo, o si es de largo pequeño, no funciona recursivo.
- MSD: funciona aún cuando los strings son de largo diference. Se ordena con CountingSort el primer caracter, y se ordenan subarreglos correspondientes a cada caracter recursivamente (excepto el primero, que es común para cada subarreglo). Para subarreglos pequeños (len()<10) combiene usar InsertionSort para los primeros k caracteres iguales.
