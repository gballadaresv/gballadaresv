# Cliente

_\* El landing page es donde llega el cliente, ahí juzga la página. De esto depende si el cliente se queda o no._

\* Hay validaciones que conviene hacerlas del lado cliente y otras del lado servidor (en el server se suele validar cosas relacionadas a dónde mete el dedo el usuario).

\* En el cliente está el primer filtro de calidad.

\* Es escencial hacer QA, pues cada error en el server repercutirá en todos los clientes.

\* CI/CD (Integración continua / Despliegue continuo): Cada vez que se hace un cambio en el código, se hacen pruebas automáticas y si todo está bien, se despliega automáticamente. Hay varias fases:
- Desarrollo: Aquí se hacen los cambios y pruebas locales.
- Pruebas: Aquí se hacen pruebas automáticas (unitarias, de integración, de sistema, de aceptación).
- Preproducción: Aquí se hace un despliegue en un entorno que simula producción.
- Producción: Aquí se hace el despliegue final.

\* Siempre hay que tener al menos 3 ambientes: Desarrollo, Pruebas y Producción. (A veces 4, con Certificación).

##### Como llega CSS y HTML al browser
```js
app.use(serve(path.join(process.cwd(), 'public')));
```
1. `process.cwd()` devuelve el directorio actual de trabajo.
2. `path.join(process.cwd(), 'public')` crea una ruta absoluta al directorio `public`.
3. `serve(...)` viene con koa-static, crea un middleware que sirve archivos estáticos desde el directorio especificado. Gestiona cabeceras útiles.
4. `app.use(...)`: registra el middleware en la cadena Koa.


##### HTML-DOM
Document Object Model

Es una interfaz que permite a los lenguajes de programación acceder y actualizar el contenido, la estructura y el estilo de un documento HTML.

\* Está organizado como un árbol de objetos, donde cada nodo representa un elemento del documento (etiquetas, atributos, texto).

##### CSSOM
CSS Object Model

Es un mapa de todos los selectores y propiedades relevantes para cada selector definido en CSS.
También se organiza en forma de árbol, pero distinto al DOM.

\* En conjunto con el DOM forman el Árbol de Renderizado (Render Tree), el "dibujo" de la página web.


##### HTML y CSS
\* La gracia de CSS es que se comunican con HTML si cumplen ciertos criterios de selección.

\* CSS: Cascade Style Sheets

Cuando HTML-DOM y CSSOM se combinan, se obtiene el render tree, un árbol de los elementos que se van a mostrar en pantalla, con sus estilos aplicados, es lo que efectivamente se renderiza en el navegador.

\* Pensar modularmente los elementos de la vista. Preguntarse qué elementos se repiten, y pensar en la estructura.

```css
selector {
    propiedad: valor;
}
```

_\* La landing page tiene que ser lo más limpia y clara posible._

##### Selectores en CSS

- Selector de tipo: selecciona todos los elementos de un tipo específico (etiqueta HTML).
```css
p {
    color: blue;
}
```

- Selector de clase: selecciona todos los elementos que tienen una clase específica.
```css
.nombre-de-clase {
    color: red;
}
```

- Selector de ID: selecciona un elemento con un ID específico (único en la página).
```css
#nombre-de-id {
    color: green;
}
```

- Selector universal: selecciona todos los elementos.
```css
* {
    margin: 0;
    padding: 0;
}
```
- Selector de atributo: selecciona elementos con un atributo específico.
```css
input[type="text"] {
    border: 1px solid black;
}
```
- Selector descendiente: selecciona elementos que son descendientes de otro elemento.
```css
div p {
    font-size: 14px;
}
```

### JS en frontend
Para seleccionar elementos, se usan métodos del DOM que devuelven una colección de todos los elementos que cumplan con el tag, clase o nombre.
```js
document.getElementById("id-del-elemento");
document.getElementsByClassName("clase-del-elemento");
document.getElementsByTagName("etiqueta-del-elemento");
```

##### Pasos para lograr controlar el entorno:
1. Estructurar el HTML: colocar div, span, main section, header, tags, etc.
2. Otorgar ids, clases y otros atributos que permitan identificar sin ambigüedad los elementos.
3. En el código de JS, crear objetos y variables adecuadas que permiten trabajar con los elementos de la página.

##### Event listeners
Es todo lo que está escuchando las interacciones con el usuario (clicks, movimientos del mouse, teclas, etc).

```js
elemento.addEventListener('click', function() {
    // Código a ejecutar cuando se hace clic en el elemento
});
```

Tipos de eventos comunes:
- `click`: cuando se hace clic en un elemento.
- `mouseover`: cuando el cursor del mouse pasa sobre un elemento.
- `mouseout`: cuando el cursor del mouse sale de un elemento.
- `keydown`: cuando se presiona una tecla.
- `keyup`: cuando se suelta una tecla.
- `submit`: cuando se envía un formulario.

\* En localStorage no guardar nada relevante como passwords, datos sensibles, etc. Solo cosas que no comprometan la seguridad del usuario (también existe sessionStorage).

### REACT

_\* Lo importante no es saber desarrollar, es saber cómo funciona, la IA puede hacer una aplicación, pero nosotros debemos entender cómo funciona para saber por qué falla._

Librería de JS para contruir UIs.
Tiene 2 elementos principales:
- React Component API.
- React DOM: encargado del rendering, toma los datos de entrada y retorna lo que debe ser desplegado en pantalla.

\* Es declarativo y basado en componentes.

\* REACT no usa el DOM real, usa un Virtual DOM. Esto permite que sea más barato actualizar la UI, pues sólo actualiza lo que cambió, no tiene que recalcular todo el DOM.

_Diseño modular, los módulos tienen módulos._

Los componentes utilizan una sintaxis llamada JSX (JS XML), es un lenguaje de marcado _muy_ similar a HTML.

\* archivos `.jsx` no son obligatorios.


`App.jsx`
```jsx
import ReactDOM from 'react-dom';

ReactDOM.render(
    <MiComponente nombre="Mundo" />,
    document.getElementById('root')
);

class MiComponente extends React.Component {
    render() {
        return (
            <div className="mi-clase"> {/* Aquí se nota que no es HTML, pues usa className en vez de class */}
                <h1>Hola, {this.props.nombre}!</h1>
                <p>Este es un componente de React.</p>
            </div> 
        );
    }
}

// props es un objeto de JS que contiene las propiedades pasadas al componente.
function Modulo(props) {
    // ...
    // código de JS
    return (
        <div className="mi-clase"> 
            <h1>Hola, {props.nombre}!</h1> {/* Todo lo que esté entre {}, se evalúa como JS */}
            <p>Este es un componente de React.</p>
        </div>
    );
}
```

\* Sólo es necesario envolver en un div si tengo más de una sentencia jsx.

React DOM se encarga de actualizar el DOM. Cuando el estado de un componente cambia, React DOM vuelve a renderizar ese componente y sus hijos, actualizando solo las partes del DOM que han cambiado.

\* El componente es reutilizable, escribo una vez, utilizo muchas.

Un componente en React tiene un ciclo de vida de 4 estados:
- Initialization: se crea el componente.
- Mounting: se inserta el componente en el DOM.
- Updating: el componente se actualiza cuando cambian sus props o estado.
- Unmounting: el componente se elimina del DOM.

##### Hooks
Los hooks son funciones que permiten a los componentes funcionales de React gestionar el estado y los efectos secundarios. En concreto sirven para "engancharse" (acceder y actualizar) al estado y al ciclo de vida de los componentes funcionales, que originalmente no tenían estas capacidades (atributos privados).
Algunos hooks comunes son:
- `useState`: permite añadir estado a un componente funcional. Puede servir para manejar datos que cambian con el tiempo, como entradas de usuario o respuestas de una API (variables).
- `useEffect`: permite gestionar efectos secundarios, como la suscripción a eventos o la obtención de datos. Sirve para renderizar algo después de que se cumpla una condición.

El dispatcher es un componente compartido que contiene todas las funciones tipo hook. Luego se guarda en memoria persistente, la primera vez que se llama un hook, se guarda el resultado, y la próxima vez que se llame, se devuelve el resultado guardado.

```jsx
import React, { useState, useEffect } from 'react';

function Contador() {
    const [state, setState] = useState(0); // Settea el valor inicial de la variable state
    // state es variable privada, setState el setter para actualizarla
    // state puede ser lo que queramos, un objeto grande, array, etc.
        useState(() => {
            // Código de inicialización
        });

    useEffect(() => {
        document.title = `Has hecho clic ${state} veces`;
    }, [state]); // Solo se ejecuta cuando 'state' cambia

    // ...
    // código de JS

    return (
        <div>
            <p>Has hecho clic {state} veces</p>
            <button onClick={() => setState(state + 1)}>
                Haz clic aquí
            </button>
        </div>
    );
} 
```

## Diseño UX/UI

Para que la gente use mi página, la gente tiene que querer usarla. Para ello hay técnicas de marketing (incluyendo uso de metadata), y un diseño atractivo y funcional (UX/UI: User Experience / User Interface):

- Cuidar el uso de colores.
- Atención a la tipografía (fuentes legibles, tamaños adecuados).
- Uso de espaciado.
- Ojo con las imágenes de fondo.
- Sitios no responsivos (adaptables a móviles) pierden muchos usuarios.
- Sin música de fondo ni pop-ups molestos.
- Exceso de "efectos especiales" puede distraer y alejar usuarios.


Recomendaciones básicas:
- No abusar de marcos (o cositas que se mueven).
- La estructura de navegación debe ser fácil (cuidad con menús desplegables).
- Fondos simples, limpios y relajantes.
- No a las animaciones (quizá alguna puntual y sutil).
- **MUCHO** ojo con la ortografía y redacción.


El contenido también incluye _contenido_ (aspectos del negocio):
- Mucho contenido que se agolpa.
- Demasiados conceptos (de qué se trata el sitio?).
- Introducciones muy largas (mucho texto.).
- Mucho scroll para llegar donde quiero.
- El contenido se mueve demasiado rápido (banners, carruseles, etc).
- Ventanas emergentes (pop-ups) que no me interesan.
- Solicitud de información personal antes de hacer el enganche.
- Mucho click para tan poco.
- Promesas incumplidas.
- "Estamos trabajando", entonces no lo muestres.
- Pq estamos acá? Pensé que era otra cosa.

_\* En el landing page está el 99% de las ventas._ Si no lo logra, en 3 segundos me voy y no entro más.

\* El landing page debe tener muy poco texto, debe ser claro y conciso. Debe responder a:
- Qué es esto?
- Para qué me sirve?
- Por qué debería importarme?
- Qué hago ahora?

##### Aspectos de UX
- Crear interfaces que sean fáciles de usar, curva de aprendizaje rápida.
- No cambiar estándares, el rojo es peligro, el verde es seguro.
- La toma de desiciones por parte del usuario debe ser una experiencia simple y fluida.
- Dejar claro el acceso a servicios.


### VueJs
Es un framework progresivo de JS para construir interfaces de usuario, se centra en la capa de vista y se basa en la idea de Single Page Applications (SPA).

\* La biblioteca es un grupo de cosas que uno puede utilizar, en cambio el framework es algo que da un espacio predefinido donde trabajar. El framework entrega muchas más cosas listas, la biblioteca requiere trabajo más detallado, pero es más flexible.

_\* Si no hay problemas, no hay ingeniería. Saber identificar el problema. La metodología y herramientas responden al problema, no el problema a la solución._

\* En VueJs no hay hooks.

\* La estructura básica es similar a React, sigue siendo JS con HTML y DOM. React usa JSX, Vue usa Templates.

\* Vue detecta los cambios automáticamente, no hace falta usar un setState o similar.

\* No se puede manejar nada de componentes en `beforeCreate()`, sólo cosas como variables globales o constantes. Para manejar componentes, usar `created()` (obviamente no puedo manejar algo que aún no existe).

_\* La IA va a beneficiar al Senior, al que sabe cosas._

Archivo `App.vue`:
```JS
<template>
   {/*  El template es un HTML con esteroides (como v-if, v-for, etc.) */}
    <div>
        <input
            :value="text"
            @input="event => text = event.target.value"
        />
        <p>Message is: {{ message }}</p> {/* Las variable de JS van entre {{ }} */}
    </div>
</template>

{/* El template desempaqueta automáticamente, el script necesita usar .value */}

<script>
    {/* código de JS */}
</script>

<style>
    {/* código de CSS */}
```

Estructura:
```plain
Name:Argument.Modifiers=Value
<input v-model="message" placeholder="edit me">

<script>
{/* código de JS */}
</script>
```

Estructura:
```plain
Name:Argument.Modifiers=Value
```
ej: 
```plain
v-on:submit.prevent="onSubmit"
```
Names:
- v-bind: vincula atributos de HTML a datos de Vue. (No se suele usar v-bind, se prefiere usar `:`)
- v-model: crea un enlace bidireccional entre datos y elementos de formulario.
- v-on: escucha eventos y ejecuta métodos.
- v-if: renderiza condicionalmente elementos.
- v-for: renderiza listas basadas en datos.
- v-show: muestra u oculta elementos basados en una condición.

Arguments:
- event: tipo de evento (click, input, submit, etc). (No se suelen usar eventos como v-on:event, se prefiere usar `@event`).
- attribute: atributo HTML (href, src, class, style, etc).
- key: clave única para listas renderizadas (usado con v-for).
- model: propiedad vinculada en un componente (usado con v-model).

Modifiers:
- .prevent: previene el comportamiento por defecto del evento.
- .stop: detiene la propagación del evento.
- .capture: usa la fase de captura del evento.
- .self: el manejador solo se activa si el evento se origina en el elemento
- .once: el manejador se activa solo una vez.

Value:
- Puede ser una expresión JS, una variable, o el nombre de un método definido en la instancia de Vue.

\* El browser va a hacer el default, más lo que uno le pide. El default puede generar sobreuso de recursos y brechas de seguridad.
Prevent default:
```JS
event.preventDefault();
```

\* El propagation es cuando un evento se "propaga" desde el elemento (div) hijo hasta los padres en el DOM. Detener la propagación evita que los manejadores de eventos en elementos padres se activen cuando un evento ocurre en un elemento hijo.
```JS
event.stopPropagation();
```

# Stack tecnológico
Cuando uno hace una aplicación web, hay muchas tecnologías. Algunas del lado del servidor, y otras del lado del cliente. El conjunto de tecnologías usadas se llama stack tecnológico.

Son desiciones de diseño de cómo voy a usar y cómo voy a hacer mi aplicación. Responde a las características del problema a solucionar.

- ¿Cuáles son las tecnologías que más se adaptan a mi problema?
- ¿Es difícil de migrar?
- ¿Open source?
- ¿Características de la DB? (velocidad, confiabilidad, escalabilidad)
- ¿Soporte de tecnologías?
- ¿Costo de operación? (licencias, requerimientos técnicos, profesionales, mantenimiento)
- ¿Curva de aprendizaje?

### JAMstack
Javascript (React, Vue, Angular), APIs (REST), Markup (HTML, CSS)

A diferencia de otros stacks (LAMP, MEAN, MERN, WAMP, PERN, etc.), JAMstack no se centra tanto en la tecnología del servidor, sino a la separación entre front y back (Js y Apis), centrándose más en la separación cliente-servidor.

\* Es complejo de ensamblar, tiene una curva más bien alta, pero es escalable.

#### Client-side rendering (CSR)
La mayoría del procesamiento y renderizado de la interfaz se realiza a capa cliente.

\* Es lo que hacemos en el curso.

#### Server-side rendering (SSR)
El procesamiento y renderizado de la interfaz se realiza en el servidor, enviando HTML ya generado, lo que se streamea al cliente.

- Si se necesitan estrategias complejas de SEO (Search Engine Optimization) para los motores de búsqueda.
- Para sitios web focalizados en el contenido, **NO** en la interacción (ej. blogs, sitios de mucho texto, más bien estáticos).
- Se cuenta con servidores de alta capacidad de procesamiento (o granja de servidores), se pueden usar balanceadores de carga y CDN (Content Delivery Network). 

\* Preguntarse si realmente se necesita, depende de la importancia para la aplicación del tiempo que se requiere para acceder al contenido.

##### Static site generation (SSG)
En vez de esperar hasta que se solicite una página y se genere la vista _on demand_, SSG lo hace de antemano, generando todas las páginas estáticas durante el proceso de _construcción_ del sitio web, con lo que la vista estará lista para cuando se haga la solicitud.

- Si el contenido no cambia con frecuencia.
- Si se busca un rendimiento óptimo y tiempos de carga rápidos.
- Para cuando se accede a la misma información muchas veces.

\* SSG es un caso particular de SSR.



_\* El stack debe elegirse según las necesidades específicas del proyecto, para optimizar rendimiento, escalabilidad y UX._
