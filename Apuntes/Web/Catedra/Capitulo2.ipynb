### Asincronía
Recordar que en JS no se ejecuta todod e forma lineal, se deja lo bloqueante de lado y se retorna en el orden que termina.

Callback es una función que llama a otra función.

##### Promesas
Es un objeto que tiene la condición de esperar hasta que termine algo.
Es la promesa que tiene el control de la asincronía, no hacia afuera.
Lo que quiero secuencializar lo encapsulo en promesas.

```python
// Sirve para ejecutar código después de un tiempo. No se pausa, se ejecuta el resto del código primero.
// timeout es bloqueante, por lo que se ejecuta en background mientras continúa el resto del código
setTimeout(() => {console.log("Hola después de 2 segundos")}, 2000);

// funcion callback, es cualquier función que llame a esta función
// se usa mucho en asincronía
function ejecutarCallback(arg, callback) {
    console.log("Antes de ejecutar el callback");
    callback(arg);
    console.log("Después de ejecutar el callback");
}

// llama a la función ejecutarCallback, pasando como argumento una función anónima (callback)
ejecutarCallback("Hola desde el callback", (arg) => {console.log(arg)});

// Promesas, son objetos que representan la terminación o el fracaso de una operación asíncrona
function promesaEjemplo(exito) {
    // la función retorna una promesa, cae en el pool de espera y se ejecuta cuando pasen cosas
    return new Promise((resolve, reject) => {
        if (exito) {
            resolve("Operación exitosa");
        } else {
            reject("Operación fallida");
        }
    });
}

// como la función con promesa queda en espera, hay que llamarla con then y catch
// notar que retorna un valor
promesaEjemplo(true)
    // el then se ejecuta si la promesa sale con resolve
    .then((mensaje) => {console.log(mensaje);})
    // el catch se ejecuta si la promesa sale con reject
    .catch((error) => {console.error(error);});

// se pueden encadenar promesas
promesaEjemplo(true)
    .then((mensaje) => {
        console.log(mensaje);
        return promesaEjemplo(false);
    })
    // notar que hay más de un then, primero se ejecuta el primero, luego el segundo
    .then((mensaje) => {console.log(mensaje);})
    // basta con que una falle para que caiga en el catch
    .catch((error) => {console.error(error);});


// promise.all espera a que todas las promesas se resuelvan
// promise.race espera a que la primera promesa se resuelva
// promise.allSettled espera a que todas las promesas se resuelvan o rechacen
// promise.any espera a que cualquiera de las promesas se resuelva

// las promesas se pueden encadenar para ejecutar de forma secuencial
function pasoA() {
    return new Promise((resolve) => {
        setTimeout(() => {
            resolve();
        }, 1000);
    });
}

function pasoB() {
    return new Promise((resolve) => {
        setTimeout(() => {
            resolve();
        }, 500);
    });
}

function pasoC() {
    return new Promise((resolve) => {
        setTimeout(() => {
            resolve();
        }, 100);
    });
}

function pasoD() {
    return new Promise((resolve) => {
        setTimeout(() => {
            resolve();
        }, 1);
    });
}
// Sin promesas encadenadas, se ejecutan al mismo tiempo, sin control del orden de ejecución
// D -> C -> B -> A

pasoA()
    .then(() => pasoB())
    .then(() => pasoC())
    .then(() => pasoD());

// Se puede ahcer lo mismo con async - await
async function main() {
    await pasoA();
    await pasoB();
    await pasoC();
    await pasoD();
}

main()
```
