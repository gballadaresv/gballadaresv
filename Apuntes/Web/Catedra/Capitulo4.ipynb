# Testing

Se espera que todo software haga lo que tenga que hacer (de forma segura y amigable), sin errores en producción.

\* El tester debe tener claro el propósito del proyecto, las reglas de negocio, y tener un gran énfasis en ciberseguridad.

\* Uno no está desarrollando por gusto para uno, está desarrollando para el usuario.

##### Análisis estático de código
Se puede hacer un análisis visual o con linters para encontrar errores, malas prácticas, problemas de sintaxis y duplicación/redundancia de código.

##### Testing unitario
Se testean módulos, funciones y métodos específicos de clases en un contexto aislado para probarlos de forma individual.

##### Testing de integración
Pruebas diseñadas para que interactúen varios componentes de software. Verifican que los componentes funciones bien en conjunto.
(ej. probar de forma integrada el módulo que realiza operaciones CRUD sobre una BD o componentes de un módulo que dependan del resultado de otro)

##### Jest
El framework de testing en JS

Algunos matchers:
- De igualdad:
  - toBe
  - toEqual
- De verdad:
  - toBeNull
  - toBeUndefined
  - toBeTruthy
  - toBeFalsy
- Números:
  - toBeGreaterThan
  - toBeGreaterThanOrEqual
  - toBeLesserThan
  - toBeLesserThanOrEqual
- Strings:
  - toMatch

```python
// Jest es el framework de testing en JS
// todo archivo con formato filename.test.js es un archivo de test

/* Agregar en package.json para ejecutar los tests con npm test
{
    "scripts": {
        "test": "jest --verbose"
    }
} 
*/

describe('test group', () => { // describe es una suite de tests
    
    beforeAll(() => {
        // código que se ejecuta una vez antes de todos los tests
    });
    afterAll(() => {
        // código que se ejecuta una vez después de todos los tests
    });
    beforeEach(() => {
        // código que se ejecuta antes de cada test
    });
    afterEach(() => {
        // código que se ejecuta después de cada test
    });


    test('test name', () => {
        expect(1 + 1).toBe(2) // expect es resultado del test, toBe es lo que esperamos (matcher)
        // expect es un callback
    });
});
```
