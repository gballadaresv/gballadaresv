La arquitectura web es cliente-servidor.

Para que sea aplicación web debe tener browser, si no tiene browser no es web (las apps de celular no son tecnologías web).
No es necesario descargar nada para utilizar, es llegar y usar.

Los componentes básicos de web son:
- Cliente.
- Browser.
- Conexión de internet (una conexión física, ej. cables submarinos).
- Servidor web.

Las 3 tecnologías fundamentales:
- HTTP

##### IP

Queremos poder acceder a un servidor de manera única, esto se logra con la dirección IP (Internet Protocol). 
IPV4 tiene la forma ###.###.###.### (4 números separados por puntos).

*localhost tiene IP 127.0.0.1 siempre (es privada).

##### TCP

El TCP (Transmission Control Protocol) es el estándar que permite establecer y mantener la conexión entre computadores en la red (define cómo está organizada la info para que el otro lado entienda).

La IP trabaja con el TCP para definir cómo se envían los paquetes de información (packet) entre computadores (TCP/IP). 

Un paquete tiene la forma: 
IP header / TCP header / data

*El dominio se puede inscribir en NIC.cl

##### DNS

DNS (Domain Name System) es un servicio que asocian un nombre con una dirección IP.

### URI - URL

Para encontrar un recurso en la web se tiene el siquiente esquema:

protocolo://subdominio.host.dominio:puerto/?query

*El protocolo se conoce como schema

*URI: identificador
*URL: localización

ej:
https://www.ejemplo.com:3000/general/articulo/detalle/main.html#page1

*NUNCA usar el puerto por defecto, no lo sé sólo yo, también lo sabe el hacker.

El RFC (Request For Comments) 3986 define la URI (Uniform Resource Identifier) que estandariza la sintaxis.

### HTTP

HyperText Tranfser Protocol es un protocolo creado para la transferencia de archivos sobre la arquitectura de internet (HTML y URIs), por lo que corre sobre TCP/IP y posee una estructura cliente-servidor.

*Ahora el estándar de seguridad es TLS 1.2+



** El frontend es el que debe ganarse al usuario, quiere una interfaz limpia, directa, suave.

** El browser trabaja con el transpiler, que trabaja en JavaScript (aunque webassembly permite permite trabajar con c, python, etc).


El HTTPS verifica que es lo que dice ser, no asegura que no hay hackers ni sea un negocio legítimo. Es una certificación (como el blue check de ig). Además asegura que está cifrado

HTTP/2 posee la característica de multiplexación, que paraleliza los proceso de envío y confirmación de recibimiento de los datos. Trabaja con TCP armando paquetes.

HTTP/3 trabaja con UDP mediante QUIC, con un stream de datos. Tiene mejor control de congestión, corrección de errores, multiplexación sin bloqueo de header.

*El 60% de los sitios usa HTTP 2, y 30% HTTP 3.

##### TCP VS UDP

TCP es altamente fiable, más lento, y asegura que la información llegue bien, mientras que UDP prioriza la velocidad por sobre la integridad de los datos.

** Cuando tenemos un ciberataque o queremos saber si un sito está activo, no hay que usar browser, hay que usar el comando _curl_:

ej: curl -I --http1.1 https://www.google.com

#### Modelo MVC
Es un patrón de diseño de software

- Modelo: el componente que maneja la lógica de negocio y datos de la aplicación (server side).
- Vista: el componente que presenta los datos al usuario (client side).
- Controlador: intermediario entre el modelo y la vista (middleware, sabe qué me están pidiendo y a quién pedírselo (mayordomo)).
