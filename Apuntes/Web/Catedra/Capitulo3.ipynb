### Hashing
Es una función que transforma datos en una secuencia.
Si hago un cambio mínimo en los datos, el Hash será completamente diferente.
Dado que los Hash son finitos y limitados, eventualmente se repetirán, por lo que se implementan _rounds_ (SALT, string único por usuario que revuelve los datos), combinaciones de los datos para generar vectores únicos para cada paso. Con esto passwords iguales, tendrán Hashes distintos, logrando que sea muy difícil desencriptar la contraseña.
El nivel de seguridad (y complejidad computacional) estará dado por la cantidad de rounds (generalmente ~12).
Bcrypt es un módulo que contiene funciones para usar Hashing.

\* El SALT se almacena DENTRO del hash generado, no por separado. Cuando se verifica una contraseña, bcrypt extrae el SALT del hash almacenado y lo usa para recrear el hash con la contraseña ingresada.

*No pretende que sea incrackeable, sólo lo hace lo suficientemente difícil para que sea muy costoso (y deje de ser rentable) hackearlo.

\* Se recomienda que len(pwd) > 8

\* Se recomienda variabilidad en la contraseña:
    "\[a-z\]\[A-Z\]\[0-9\]\[symbol\]"

\* ej del hash: \$2b\$salt$hash
<!-- salt son 22 caracteres, hash 31. -->



1. Generación de salt automático.
2. Combinar clave + salt y aplicar hashing varias veces (rounds configurables).
3. Almacenar el resultado que incluye: versión + rounds + salt + hash.
4. Para verificar: extraer salt del hash almacenado y recrear el proceso.


\* No necesariamente hashing es para cifrado, también se puede usar para generar Tokens (JWT).

```python
const bycript = require('bcrypt');

(async () => {
    // definimos nuestro algoritmo de hashing
    const COSTO = 12;
    const salt = await bycript.genSalt(COSTO);

    // almacenamos la contraseña hasheada
    let stored_pwd = 'examplePassword123';
    const hashed_stored = await bycript.hash(stored_pwd, salt);

    // introducimos la contraseña del usuario
    let input_pwd = 'examplePassword123';

    // comparamos la contraseña introducida con la almacenada
    const compare = await bycript.compare(input_pwd, hashed_stored);
})(); // (); ejecuta la función inmediatamente (1 vez)
```
