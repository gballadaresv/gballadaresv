- XSS (Cross Site Scripting) reflejado:
1. Usuario inyecta código malicioso en un sitio web.
2. El servidor lo devuelve.
3. El navegador lo ejecuta.

- XSS almacenado: el ataque queda guardado en los datos persistentes, el navegador lo ejecuta porque confía en todo lo que viene del servidor.

NIST es un framework de ciberseguridad:
- Protect: controles técnicos de seguridad.
- Detect: monitoreo de tráfico.
- Identify: endpoints críticos.

CWE (Common Weakness Enumeration) es un catálogo de debilidades comunes de software.

OWASP (Open Web Application Security Project) es una comunidad online que crea artículos, metodologías, documentación, herramientas y tecnologías en el campo de la seguridad de aplicaciones web.

### Inyección SQL
El atacante inserta código SQL malicioso en una consulta SQL a través de la entrada de datos del cliente.

ej:
Cuando en la URL es: `www.ejemplo.com/producto?id=5`, el atacante puede modificar el parámetro `id` para inyectar código SQL malicioso. Por ejemplo, cambiando la URL a `www.ejemplo.com/producto?id=5 OR 1=1`, el atacante podría manipular la consulta SQL para devolver todos los productos en lugar de solo el producto con ID 5.

Para prevenir inyecciones SQL, se pueden utilizar consultas preparadas (prepared statements) y procedimientos almacenados (stored procedures), que separan los datos de la lógica SQL, evitando que el código malicioso sea ejecutado.

### Seguridad de cabeceras
- Content Security Policy (CSP): ayuda a prevenir ataques _XSS_ al especificar qué fuentes de contenido son confiables.
- Cross-Origin Resource Sharing (CORS): controla qué dominios pueden acceder a recursos en un servidor, ayudando a prevenir ataques de tipo cross-site.
- X-Frame-Options: previene ataques de _clickjacking_ al controlar si una página puede ser cargada dentro de un iframe.
- X-Content-Type-Options: previene ataques de _MIME sniffing_ al indicar que el navegador debe respetar el tipo de contenido declarado.

\* Koa-helmet es un middleware para Koa que ayuda a proteger aplicaciones web estableciendo varias cabeceras HTTP de seguridad.

```JS
const Koa = require('koa');
const helmet = require('koa-helmet');
const app = new Koa();
app.use(
    helmet.contentSecurityPolicy({
        directives: {
            defaultSrc: ["'self'"],
            scriptSrc: ["'self'", 'trusted-scripts.com'],
            imgSrc: ["'none'"],
        },
    })
);
```

### CORS
La biblioteca de CORS (Cross-Origin Resource Sharing) (ej. `@koa/cors`) es un middleware que implementa la lógica necesaria para controlar las peticiones cross-origin en apps web. Controla qué navegadores pueden hacer peticiones desde _otros dominios_ hacia el backend.

ej:
```Plain
Frontend: http://localhost:5500
Backend: http://localhost:3000
Si el front intenta hacer fetch("http://localhost/api/data"), el navegador bloque la petición (aunque el backend responda), a menos que el backend envíe cabeceras CORS correctas.
```

Las cabeceras mínimas para permitirlo:
- Access-Control-Allow-Origin: http://localhost:5500
- Access-Control-Allow-Methods: GET, POST
- Access-Control-Allow-Headers: Content-Type
