2. Escribir wsl en la terminal

Comandos de wsl:
git pull: baja desde la nube a local
git push: sube los cambios a la nube

Para subir archivos: 
1. git add {archivo}
2. git commit -m "{mensaje}"
3. git push
*Usar git status de vez en cuando

pwd: muestra el path al directorio actual
wslview . abre una visualización de la carpeta
touch: crea un archivo


getter: permite obtener atributos privados

@Property
def atributo(self):
    return self.__atributo

setter: permite modificar un atributo priuvado, se usa despues del setter

@atributo.setter
def atributo(self, cambio)
    self.__atributo = cambio

deleter: elimina un atributo, se usa despues del setter
*despues en el codio se usa del

@atributo.deleter(self):
    del self.__atributo

del instancia.atributo


El super se usa sin self
