Patrones de diseño:

Contexto: Determina la estrategia a utilizar

Estrategia: Cada estrategia es una clase que ejecuta una serie de métodos según lo que se requiera.

### Strategy
El contexto es una clase que contiene como atributos privados las estrategias, y de métodos (públicos) un método que settee la estrategia.

Cada estrategia es una clase que tiene sus propios atributos y métodos

### Composite
Tiene un clase base que sirve de template, y clases hojas (o ramas) que se encargan de toda la lógica

La dependencia no guarda una instancia de otra clase como atributo, sólo la utiliza, por ejemplo en un método.
Nótese que una clase puede crear a otra, pero si no es la única que la guarda, entonces sería dependencia, no composición.
Composición no necesariamente el que los crea, sino el que los almancena (ver qué pasa si se elimina).
Las relaciones simple son semánticamente iguales a una agregación en ruby

En un diagrama de secuencia, los métodos van hacia la línea de vida, que se representan con una línea punteada transversal al llamado de los distintos métodos.

No todas las clases se crean simultáneamente al inicio de la ejecución del código, alguas otras clases se crean dentro de la ejecución del código, en tal caso se representan con la clase más abajo en la secuencia y un mensaje de creación (línea punteada).

Los bucles también se representan con bloques

Si una clase se llama a sí misma se puede agregar una barra de acticación encima de la otra.

Para la composición en ruby se pueden usar variables de instancia (@) para guardarlas.

Agregación recibe una instancia ya creada de otra clase.

La dependencia no guarda una instancia, sólo la usa.

Para hacer una relación simple consigo mismo, puede ser un método que reciba una instancia de objeto de la misma clase y haga algo con él.

*En ruby se pueden definir métodos de clase usando el self. (def self.metodo)

#### Diseño
### Observer
Un publisher envía mensajes, tiene de atributo una lista que tiene lso nombres de los subscribers. Los subscribers pueden ser distintas plataformas de notificación (ej, whatsapp, telegram). El publisher tiene un método para agregar observers, y uno para notificar a los observers

### Strategy
Se crea un selector de estrategia, similar a una clase abstracta, pero no igual. Una vez elegida la estrategia, se define cada estrategia que funciona por separado.

### Template
Se crea un template base que será heredado, y luego se van sobreescribiendo los métodos según se requiera.


#### Creacionales
### Singleton
Se crea una única instancia de una clase, se verifica que ya existe la instancia de la clase, y sólo si no existe se crea la isntancia (el méodo de creación es privado para que no se pueda acceder de forma externa)

### Builder
Es una clase que crea todo

### Factory
Una fábrica que crea distintos objetos según se requiera

### Abstract factory
Es como crear fábricas de fábricas, modulariza la creación de objetos con la misma estructura.

#### Estructurales
### Adapter
Se crea un paso intermedio para adaptar la información al código, en vez de adaptar el código a la información.

### Proxy
Va controlando el acceso a distintas partes del código según los permisos de un usuario determinado.

### Decorator
Es como agregar un wrap sobre la pintura, se añade un decorador para cambiar el objeto de retorno, vamos agregando "capas".

### Composite
Se parte de una raíz, y se van creando las hojas  (hijos), son las hojas las que ejecutan todo.
