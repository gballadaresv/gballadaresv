# Enfoques y escuelas de la gestión

### Administración científica
Taylor

Postula usar métodos científicos para definir la forma de realizar un trabajo, establece un método estandarizado para realizar el trabajo. Busca optimizar la ejecución de tareas, asignar el trabajo a la persona correcta con las herramientas correctas.
Asume que los trabajadores no tienen la capacidad crítica para optimizar el proceso, debe ser el superior que da las órdenes.

### Teoría general de la administración
Fayol y Weber

14 Principios de Fayol:
- División del trabajo: especialización aumenta el rendimiento
- Autoridad: los gerentes tienen el derecho a dar órdenes
- Disciplina: los empleados deben obedecer y respetar las reglas
- One boss: todo empleado debe recibir órdenes de un solo superior
- Goal alignment: debe haber un solo plan de acción
- Interés general sobre individuales.
- Remuneración: los empleados deben recibir un pago justo
- Centralización: el grado en que los empleados se involucran en la toma de deciciones
- Escalafón: línea de autoridad desde el gerente de nivel más alto a los niveles inferiores
- Orden: las personas y materiales deben estar en el lugar y momento adecuados
- Equidad: los gerentes deben ser amables y justos con los empleados
- Estabilidad en los puestos del personal: debe haber una planeación de personal ordenada y garantizar que hay reemplazos disponibles para las vacantes.
- Iniciativa: los empleados que tengan permitido originar y ejecutar planes, tendrán mayor nivel de esfuerzo.
- Espíritu de equipo.

### Enfoque cuantitativo
Investigación de operaciones, six sigma

La toma de desiciones se hace desde análisis estadísticos, modelos matemáticos y simulaciones computacionales.

### Enfoque conductual
Elton Mayo (experimento de la iluminación)

Los trabajadores responden al interés y atención que se pone sobre sus necesidades.
Los gerentes obtendrán mejores resultados si les dan énfasis a las personas.
El foco debería estar en la motivación, liderazgo, trabajo en equipo y manejo en conflictos.

# Estructura y diseño organizacional

La estructura organizacional es la distribución formal de los puestos de una organización.

### Diseño organizacional
##### Especialización del trabajo
El grado en que las tareas de una organización se dividen en trabahos separados.
La especialización permite un aumento en la eficiencia.
Una sobreesoecialización implica aburrimiento, fatiga, estrés, etc.

##### Departamentalización

- Funcional: Agrupa los puestos según las funciones.
- Geográfica: Generalmente se dividen sólo algunas áreas
- Por productos
- Por clientes

##### Cadena de mando
Especifica quién le reporta a quién. Puede ser más rígida o flexible.

- Autoridad de línea: Posición de autoridad que dirige el trabajo de los empleados operativos.

- Autiridad de staff: Tienen cierta autoridad, pero son más para apoyar y asesorara las personas que tienen autoridad de línea.

##### Tramo de control
Establece qué tanta gente dirigen los gerentes. Mientras meno tramo de control, menos gente tendrá a cargo cada gerente, esto permite un mayor control, pero requiere tener más gerentes.
En una organización más plana, cada gerente está a cargo de más personas, lo que disminuye la cantidad de gerentes necesarios, haciendo a la organización más horizontal (sin tanto nivel intermedio). Permite mayor autonomía a los trabajadores.

##### Centralización
El grado en que las desiciones se toman en los niveles superiores de la organización. En organizaciones centralizadas los gerentes de alto nivel tienden a tomar las decisiones, y los empleados de más bajo nivel simplemente las cumplen.

En organizaciones más descentralizadas se permite que los gerentes de niveles inferiores (que están más cerca de la operación) tomen desiciones.

##### Formalización 
El grado en que los trabajadores y procedimientos están estandarizados.
Baja formalización implica menos restricciones en la forma de trabajar de los empleados


#### Organizaciones mecanicistas
Alta especialización, departamentalización rígida, clara cadena de mando, tramos de control limitado (micromanagement?), centralización, gran formalización.

#### Organizaciones orgánicas
Equipos interfuncionales y multijerárquicos, libre flujo de información, grandes tramos de control, descentralización, poca formalización.

##### Factores de contingencia:
- Estrategia: _la estructura sigue a la estrategia_.
    - Innovación empresarial: (diferenciación) aquellas organizaciones que buscan una ventaja competitiva mediante innovaciones suelen preferir estructuras orgánicas.
    - Minimización de costos: controlar estrictamente los costos suelen tener asociada una estructura más mecánica.

- Tamaño: a medida que una organización crece, suele cambiar de una estructura orgánica a una más mecanicista con mayor especialización, departamentalización, centralización, reglas y normas.

- Incertidumbre del entorno: 
    - Las estructuras mecanicistas suelen ser más eficaces en entornos estables y sencillos.
    - Las estructuras orgánicas se adaptan mejor a ambiented dinámicos y complejos.

#### Diseños organizacionales tradicionales
##### Estructura simple
Suelen ser empresas más pequeñas o que están partiendo. Suelen tener gran centralización pero baja departamentalización, amplios tramos de control y poca formalización.

##### Estructura funcional
Alta departamentalización, se agrupan las especialidades ocupacionales similares o relacionadas.

##### Estructura divisional
Formada por unidades de negocio separadas con autonomía relativamente limitada por la coordinación y control de la matriz corporativa.

_La mayoría de los emprendimientos comienzan con una estructura simple, y a medida que crecen migran hacia diseños funcionales o divisionales._

#### Diseños organizacionales contemporáneos
##### Estructura basada en equipos
La organización está formada por grupos de trabajo o equipos autodirigidos a los que se le otorga poder de desición.
(ej. el equipo de producción de una película)

##### Estructura matricial
Se asignan especialistas de distintos departamentos funcionales para que trabajen en proyectos dirigidos por un gerente de proyecto.
Las personas de distintas áreas se combinan en proyectos.
ej. Alguien de marketing que participa en un proyecto, tiene su jefe del área de marketing, y su jefe de proyecto.
(Tribus en BICE)


# Fundamentos de planificación
Planificar implica definir los objetivos de la organización y desarollar planes en todos los niveles organizacionales (qué hacer y cómo hacerlo).

Proporcionan dirección a gerentes y empleados, reduce la incertidumbre, minimiza el desperdicio y redundancia, y establece objetivos que se pueden seguir y controlar.

Los objetivos proporcionan dirección y criterios para evaluar el desempeño.

Los planes describen cómo se lograrán los objetivos. Describen las asignaciones de recursos y establecen programas de actividades.

#### Tipos de planes
Se determinan según:
- Alcance (estratégico/operacional)
- Marco de tiempo (largo/corto plazo)
- Especialidad (direccional/concreto)
- Frecuencia de uso (uso único/permanente)

##### Alcance
- Planes estratégicos: Aplican a toda la organización, involucran al menos establecer obgetivos generales y estrategias. Abarcan perdíodos de tiempo más largos. Buscan posicionar a la organización en su entorno. Misión y visión.


#### Establecimiento de objetivos
##### Top-down
Tradicional. Los objetivos se definen desde la alta dirección. 

##### MBO 
Management by objectives. Los empleados y gerentes determinan conjuntamente los objetivos. Deben ser SMART.

##### SMART
Specific: debe ser de algo concreto.
Measurable: se debe poder medir.
Attainable: deben ser desafiantes pero posible.
Relevant: debe aportar al objetivo.
Time-bound: debe tener un horizonte de tiempo.


# Planificación estratégica

##### Misión
Es una declaración del propósito de la organización. Provee de identidad.

##### Visión
Imagen que la organización tiene de sí misma en el futuro, es una fuente de inspiración para la organización.

##### proceso de administración estratégica
1. Identificar misión de la organización, sus objetivos y estrategias.
2. Reealizar un análisis(FODA/SWOT) externo (oportunidades y amenazas) e interno (fortalezas y debilidades).
3. Formulación de etrategias
4. Implementación de estrategias
5. Evaluación de resultados

### Estrategias Corporativas
Son el plan general de los gerentes y la organización, especifica en cuáles negocios entre una compañía. Se basan en la misión y objetivos de la organización, especifica cómo cada unidad ayudará en su cumplimiento.

Tipos:
- Crecimiento: expansión a nuevos productos y mercados.
- Estabilidad: mantenimiento del estado actual.
- Renovación: examen y tratamiento de las debilidades de la organización que están provocando una caída en el rendimiento.

##### Estrategia de crecimiento
Estrategia corporativa que se usa cuando una organización desea expandir el número de mercados a los que sirve o productos que ofrece, ya sea a través de sus negocios actuales o negocios nuevos.

- Concentración: enfocarse en una línea de negocio primordial e incrementar el número de productos ofertados.

- Integración vertical: suele darse por costos o ineficiencias en los procesos externos.
    * hacia atrás: intenta obtener el control de sus entradas y ser su propio proveedor.
    * hacia adelante: intenta obtener el control de sus salidas, controlar los canales de distribución y convertirse en su propio distribuidor.

- Integración horizontal: combinar operaciones (adquirir o fusionar) con otros competidores de la misma industria para aumentar las fortalezas competitivas con otros rivales de la industria (ej. LAN + TAM = LATAM). Suelen ser observadas de serca por entidades como la FNE y el tribunal de libre competencia.

- Diversificación no relacionada: crecer al combinarse con industrias no relacionadas con baja correlación (diversificación de portafolio)

##### Estrategia de renovación
Estrategia corporativa para detener la caída del desempeño organizacional.

- De reducción: de corta duración para problemas menores y puntuales del desempeño.
- De recuperacion/reestructuración: trata problemas críticos a largo plazo mediante reducción de costos y soluciones a gran escala, usualmente requiriendo reestructuración de la empresa.

### Estrategias competitivas
Pretenden remarcar o generar el sello distintivo de la organización, que la separa de otras (ventaja competitiva).

Dependen de las nuevas entradas, compradores, proveedores, sustitutos y la intensidad de la rivalidad entre competidores actuales.

- Estrategia de liderazgo en costos.
- Estrategia de diferenciación.
- Estrategia de enfoque.

##### Estrategia de liderazgo en costos
Trata de tener los menores costos posibles. Se hace todo lo posible para recortar gastos. Los productos deben ser comparables en calidad en relación a sus rivales.

##### Estrategia de diferenciación
Tratan de crear un producto o servicio único por el que los clientes paguen más. Las diferencias pueden provenir de calidad superior, servicio extraordinario, diseño innovador, capacidad tecnológica o una imagen de marca inusuakmente positiva.

##### Estrategia de enfoque
De segmento o nicho. Incorpora el enfoque de una de las dos pero enfocada en explotar un segmento particular (nicho) del mercado. Los segmentos pueden estar basador en una gran variedad de productos (ej. cerveza artesanal), un tipo particular de consumidor (ej. fan de deportes), canal de distribución (ej. solo online y delivery), o una ubicación geográfica.

### Estrategia Portiana vs Blue ocean
- Portiana: (modelo de las 5 entradas, referir estrategias competitivas) son aplicables a mundos menos dinámicos, más predecibles, con configuraciones monopólicas perdurables y orientadas al poder.
- Blue water: Mayor incertidumbre, mayor velocidad, mayor información, con una orientación al valor (nichos no identificados, con alto potencial de desarrollo y crecimiento). Ofrecen un valor nuevo y superior a sus clientes como principal estrategia.

# Redes sociales

### Representaciones
- Grafos, nodos y arcos
- Matriz de adyacencia

### Métricas de redes sociales
- Densidad: Relaciones existentes / potenciales. Va de 0 a 1
- Reciprocidad
- Centralidad
    * Degree: n° de relaciones de un nodo, captura la probabilidad de estar expuesto a lo que fluye por la red. En grafos dirigidos hay indegree y outdegree. 
    * Closeness: distancia promedio de un nodo a otro, captura la velocidad de exposición de lo que fluye por la red.
    * Betweeness: frecuencia con la que un nodo se encuentra en el camino más corto entre 2 nodos, captura el potencial de controlar lo que fluye por la red.

### Teoría de redes sociales
- Homofilia: similares se relacionan con similares, induce clustering.
- Clique: grupo donde todos están conectados con todos
- Triadic closure: si 2 nodos tienen un amigo en común, es más probable que ellos tmb se amiguen, induce clustering.
- Network closure: las redes tienden a completar conexiones mediante cierres, termina produciendo aglutinación y cliques.
- Brokerage: intermediación, es la conexión entre grupos, los intermediarios (brokers) son raros y valiosos. Relacionan cliques.
- Small worlds: en toda red social, dos personas estarán conectados por un número pequeño (~6) intermediarios. Resulta de la combinación de muchas conexiones fuertes y algunas débiles. Lo que fluye por la red se difunde rápido, si un nodo falla no se desconecta la red.

*Una formación estratégica de links que unan cliques permite controlar flujos de información y recursos.

# Límites de la racionalidad en la toma de desiciones

### 4 Funciones de la administración
- Planificar: OKR, estrategia, plan de acción.
- Dirigir: cómo voy a motivar a las personas y resolver conflictos.
- Organizar: cómo voy a estructurar a la empresa.
- Controlar: ajustes de actividades, personas y sistemas.

### Métodos para la toma de desiciones

##### Modelo racional
Postula que los gerentes toman desiciones consistentes que maximicen el valor.

Supone que los tomadores de decisiones son perfectamente racionales, absolutamente objetivos y lógicos, se han definido cuidadosamente el problema y considerado todas las alternativas, con un objetivo claro y específico.

1. Definir el problema
2. Identificar criterios de desición
3. Asignar pesos o ponderación a los criterios
4. Desarrollar las alternativas
5. Evaluar las alternativas
6. Seleccionar la _mejor_ alternativa.

*Supone información perfecta, todas las alternativas conocidas y maximizar la utilidad (óptimo global).

##### Racionalidad acotada
Postula que la toma de desiciones es racional, pero limitada por la capacidad de un individuo de procesar información.

Supone que los tomadores de desiciones no buscarán o tendrán conocimiento de todas las alternativas, y que buscan satisfacer en vez de optimizar el resultado (óptimo local).

*Es un modelo más realista.



# Sesgos y heurísticas en la toma de desiciones

##### Heurísticas
Reglas empíricas que usan los gerentes para simplificar la toma de desiciones. Están afectas a sesgos cognitivos y de percepción.

### Sesgos y errores

##### De mérito y culpa
- Error de atribución fundamental: Se culpa a la persona en vez de la situación, se subestima la influencia de factores externos, y sobreestimar la de factores internos.
- Sesgo de auto-servicio: El éxito es "nuestro" y el fracaso "ajeno".

##### Simplificaciones
- Percepción selectiva: se interpreta selectivamente en base a intereses propios, background y experiencias previas.
- Efecto halo: formarse una impresión general de un individuo basado en una sola característica (juzgar el libro x su portada).
- Efecto contraste: evaluación de las características de una persona se ve afectada por comparaciones con otras personas que se hayan visto recientemente.
- Estereotipar: juzgar a alguien en base a la percepción que se tiene del grupo al que pertenece.
- Perfilar: forma particular de estereotipado donde miembros de un grupo son singularizados para juzgarlos intensamente basándose en un sólo razgo (raza, etnia, nacionalidad, etc.).

##### Sesgos en vida organizacional

- Exceso de confianza: cuando piensan saber más de lo que saben, o sobreestiman su capacidad de tomar buenas decisiones, especialmente fuera del área de expertise.
- Gratificación inmediata: querer obtener resultados inmediatos y evitar costos, aún cuando los resultados a futuro puedan ser superiores.

##### Sesgos en toma de decisiones
- Anclaje: quedarse con la información inicial para hacer juicios y tomar desiciones de tal forma que se obstruyen actualizaciones (quedarse con la primera impresión).
- Confirmación: buscar y aceptar fácilmente información que reafirme nuestros jucios anteriores, y criticar, rechazar e ignorar información que lo contradiga (caso particular de percepción selectiva).
- Disponibilidad: se tiende a dar mayor importancia a eventos más recientes.
- Falacia del costo hundido: se incrementa el compromiso con una desición en la que ya se ha invertido, aunque se sepa que ya no es la mejor.
- Aleatoriedad: querer dar sentido causal a eventos aleatorios y extraer significado de éstos (ej. supersticiones).
- Retrospectiva: creer que se podría haber predicho el resultado de un evento, luego de saber el resultado.


\* Gran parte de la toma de desiciones se caracteriza por racionalidad acotada, sesgos y errores de juicio, y uso de la intuición.

##### Sesgos y errores en situaciones de la vida organizacional
- Entrevistas de trabajo: sesgos perceptuales de entrevistadores afectan la precisión de los juicios sobre los postulantes.
- Expectativas de desempeño: un rendimiento alto o bajo de empleados refleja ideas preconcebidas de los líderes sobre las capacidades de los empleados (profecía autocumplida. Si creo que puedo, tengo razón, y si creo que no puedo, también tengo razón).
- Evaluación del desempeño: suelen estar influenciadas fuertemente por la percepción subjetiva de los evaluadores.

\* Para negociar sueldo: averiguar el rango de sueldo en el área, rango y puesto. Dentro de ese rango, pedir el más alto, esto permite _anclarse_ a un punto de partida alto, y negociar de ahí hacia abajo.

# Conflicto, negociación y persuasión

### Conflicto

Es un proceso que comienza cuando cuando un individuo percibe que otro ha afectado (o está por afectar) negativamente algo que le interesa.

##### Enfoque tradicional
Postula que todo conflicto es dañino y debe ser evitado.
Resulta de mala comunicación, falta de apertura y confianza, y falla de los directivos en su responsabilidad con las necesidades de los empleados.

##### Enfoque de las relaciones humanas
Plantea que el conflicto es un fenómeno natural e inevitable en cualquier grupo.
No es necesariamente dañino, tiene el potencial de ser una fuerza positiva en el desempeño grupal.

##### Enfoque interaccionista
Plantea que el conflicto no sólo es una fuerza positiva, sino que es necesaria para un desempeño efectivo.
Promueve que haya conflictos, pues en un grupo armonioso, pacífico y tranquilo se puede volver estático, apático y sin capacidad de innovar.
Incentiva a los líderes a manetener un nivel mínimo de conflicto, suficiente para que el grupo se conserve como viable, autocrítico y creativo.

- Funcional: constructivo, ayuda a mejorar el desempeño del grupo.
- Disfuncional: destructivo, daña el desempeño del grupo.

- Conflicto de tarea (task conflict): sobre el contenido y los objetivos del trabajo. Niveles bajos-moderados son funcionales. Mejora el procesamiento de información, posibilita mejor disposición al cambio y más innovación.
- Conflicto de relación (affective/emotional conflict): basado en relaciones interpersonales y desacuerdos emocionales. Se caracterizan por la ira y hostilidad. Siempre son disfuncionales. Disminuye el compartir información y cae la productividad, hace aumentar la rotación (turnover).
- Conflicto de proceso: sobre cómo debe hacerse el trabajo. Niveles bajos son funcionales.

\* En general los grupos más homogéneos tienden a tener menos conflicto.
\* Mientras más diverso el grupo, más conflictos habrán.

Es difícil mantener un conflicto de tarea sin que se transforme en uno afectivo. La gente suele interpretar los conflictos como ataques personales.

\* Con mayor diversidad en los equipos, es más probable una mayor innovación (breaj¿through), pero también es más posible que se haga un aporte insignificante 

##### Estrategias para manejar el conflicto
- Competencia: tratar de ganar a expensas del otro. (alta asertividad, baja cooperación)
- Evitación: no confrontar el conflicto, ni tratar de resolverlo. (baja asertividad, baja cooperación)
- Acomodación: ceder ante las demandas del otro. (baja asertividad, alta cooperación)
- Colaboración: ambas partes trabajan juntas para maximizar los resultados. (alta asertividad, alta cooperación)
- Compromiso: cada parte cede algo y gana algo. (moderada asertividad, moderada cooperación)

### Negociación
Proceso en el que dos o más partes intercambian bienes o servicios y tratan de ponerse de acuerdo en el valor de aquello que intercambian.
Estrategias:

- Distributiva: se da cuando hay algo que repartir, si yo gano, el otro pierde.
  - Hacer la primera oferta, y que sea agresiva (anclaje): una vez establecido el punto de anclaje, se suele tratar de ajustar en base al ancla, se favorece mucho a la persona que hace la primera oferta. Se recomienda pedir lo más alto que la otra persona podría ofrecer razonablemente (upper bound).
  - Establecer un deadline.
\* Suele usarse para negociaciones puntuales, alguien siempre sentirá que perdió, puede quedar resentida.

- Integrativa/colaborativa: buscar una solución win-win.
  - Negociar en equipos: se generan más ideas cuando hay más personas.
  - Poner más temas sobre la mesa: con más temas hay más opciones de encontrar un win-win.
  - No ceder con facilidad: si una parte cede con facilidad no se requiere creatividad para llegar a un acuerdo win-win.
\* Suele usarse en negociaciones de largo plazo, donde ambas partes quieren mantener una buena relación.


A veces las partes no logran ponerse de acuerdo a través de negociación directa, y recurren a un tercero neutral (mediador o árbitro):
- Mediador: facilita una solución negociada a través del razonamiento. Es altamente eficaz para niveles de conflicto intermedio.
- Árbitro: tiene la autoridad para dictar un acuerdo.
- Consultor: está entrenado para el manejo de conflictos, no pretende arreglar la disputa, sino mejorar la relación para que las partes puedan llegar a un acuerdo por ellas mismas.

# Poder

_\* No necesariamente se debe tener un puesto formal en una organización para tener poder._

##### Poder formal
Proviene de la capacidad para obligar o compensar desde la autoridad formal.

- Poder coercitivo: depende del miedo, se basa en la aplicación o amenaza de sanciones si el individuo no logra los resultados esperados. _A_ tiene poder coercitivo sobre _B_ si lo puede despedir, suspender, degradar, asignar actividades desagradables, etc.
- Poder de recompensa: se basa en la capacidad de distribuir premios que otros consideran valiosos, las personas cumplen porque les produce beneficios. Pueden ser recompensas financieras (nivel salarial, bonos) o no financieras (reconocimiento, asignación de trabajo interesante, turnos de trabajo preferentes).
- Poder legítimo: es el resultado del puesto en la jerarquía formal de la organización, requiere que otros miembros acepten la autoridad del puesto.

##### Poder personal
Proviene de las características personales únicas del individuo

- Poder de experto: influencia que se basa en aptitudes, conocimientos especiales o experiencia.
- Poder referente (influencers): se basa en la posesión del individuo de características personales deseables, se desarrolla por la admiración del otro y el deseo de ser como esa persona (si quiero ser como tú, hago lo que tú haces, porque quiero agradarte).

##### Poder estructural
Proviene de la posición del individuo en la red social de la organización.
_No es lo que sabes, sino a quién conoces._

Métricas de centralidad:
* conectividad (degree): número de otros a quienes un individuo está conectado directamente.
* cercanía (closeness): habilidad de llegar a otros individuos a través de un número mínimo de conexiones.
* intermediación (betweeness): grado en el que un individuo intermedia (broker) las conexiones entre otros individuos.

### Estatus
Grado en que un individuo (o grupo) es respetado o admirado por otros.
Existe enteramente a ojos de otros.

\* Las personas con bajo estatus y alto poder, tienden a asignar tareas más humillantes o denigrantes para tratar compensar su bajo estatus y bajar el de otros.

# Política
Actividades para adquirir, desarrollar y usar poder para influir en la toma de desiciones de otros.
Es el resultado de la escasez de recursos, recursos limitados llevan a conflicto y comportamientos competitivos y políticos.

##### Ejemplos de comportamiento político
- Formar bandos (coaliciones).
- Delatar o esparcir rumores (presión).
- Intercambiar favores.
- Lobby a favor o en contra de un individuo o propuesta.
- Retener información clave para quienes toman desiciones.

##### Comportamiento político usualmente aceptado (legítimo)
- Quejarse con un supervisor.
- Saltarse la cadena de mando.
- Formar bandos.
- Obstruir políticas o desiciones organizacionales.

##### Comportamiento político controversial
- Sabotaje.
- Delación (whistleblowing).
- Protestas simbólicas.
\* Las formas extremas de comportamiento político son riesgosas, pueden llevar a sanciones extremas o quedar sin poder.


\* Los "hechos" en una organización rara vez son objetivos, suelen estar a interpretación, usándose como data para tomar desiciones.

Los juicios sobre una acción difieren según la percepción del observador.
Ej:
- _Asignar responsabilidad_ o culpar a otros?
- _Documentar las desiciones_ o cubrirse la espalda?
- _Inspirar el cambio e innovación_ o amenazar el statu quo?
- _Facilitar el trabajo en equipo_ o formar bandos?

La mayoría de empleados tienen poca voluntad de participar en el juego político, y suelen tener efectos negativos:
- Disminución de la satisfacción laboral.
- Aumento de la ansiedad y estrés.
- Se eleva la rotación de personal.
- Disminuye el desempeño.

#### Gestión de la impresión
Proceso por el cual los individuos tratan de controlar la impresión que otros tienen de ellos.
Se basa en regular y controlar información en interacciones sociales en función de sus objetivos.
- La gente se comporta como si fueran actores de una obra de teatro, y los demás la audiencia.
- La interacción social es una actuación (performance).
- Se sigue el guión del rol que están jugando, y usan vestimenta apropiada al rol. Poca gente tiene acceso al backstage (lo que realmente piensan y sienten).

Técnicas:
- Auto-promoción: resaltar las mejores cualidades propias, minimizar las desventajas y llamar la atención de los logros. **Muy** efectivo en entrevistas de trabajo, contraproducente en evaluaciones de desempeño.
- Flattering: hacer cumplidos a otros respecto de sus virtudes por parecer más perceptivos y agradables. Moderadamente efectivo en entrevistas de trabado, **muy** efectivo en evaluaciones de desempeño.
- Asociación: mejorar la imagen propia al asociarse con personas, eventos o cosas que son valoradas positivamente por otros (ej. mencionar que se trabaja con un cliente importante, o que se pertenece a una organización prestigiosa).

# Liderazgo

Existen diversas teorías y definiciones de liderazgo. Una definición es la "capacidad de influir en un grupo para el logro de una visión o conjunto de metas".

### Teorías basadas en personalidad
Popular en los años 20 y 30. Plantean que los líderes nacen con una serie de rasgos o características que los diferencian del resto.

Rasgos asociados:
- Impulso: muestran un alto nivel de desempeño y tienen un deseo de logros alto, son ambiciosos, tienen energía y son perseverantes.
- Deseo de dirigir: tienen un fuerte deseo de influir y guiar a otros, demuestran disposición a tomar responsabilidades.
- Honestidad e integridad: construyen relaciones de confianza con sus seguidores, muestran congruencia entre lo que dicen y hacen.
- Confianza en sí mismos: demuestran confianza para demostrar la rectitud de sus objetivos y desiciones.
- Inteligencia: necesitan ser lo suficientemente inteligentes para reunir, sintetizar e interpretar gran cantidad de información, así como crear opiniones, resolver problemas y tomar desiciones acertadas.
- Conocimiento relativo al trabajo: el conocimiento profundo permite que se tomen desiciones bien informadas, y que se entienden las implicaciones de las mismas.
- Extroversión: son personas enérgicas y vivaces, sociables y asertivos.

Hoy se considera una teoría obsoleta, tener los rasgos asociados al liderazgo no implican liderazgo efectivo.

### Teorías basadas en el comportamiento
El liderazgo es una habilidad y se puede enseñar a cualquiera. Se deben identificar los comportamientos apropiados para enseñar a líderes potenciales.

- Orientado a la producción: enfatiza los aspectos técnicos del trabajo y logro de las tareas.
- Orientada a los empleados: enfatiza las relaciones interpersonales.

\* Los líderes orientados a los empleados logran niveles más altos de desempeño del grupo y satisfacción laboral.

No se logra un consenso de cuál es la combinación de comportamientos más efectiva.

### Teorías de contingencia

Postula que es el entorno en el que desarrolla el liderazgo.

- Modelo de Fiedler: sostiene que el desempeño del grupo depende de la coincidencia entre el estilo del líder, y el grado en que la **situación** le da control.
  - Un estilo de liderazgo determinado es más efectivo en ciertas situaciones que en otras.
  - Los líderes no cambian fácilmente su estilo de liderazgo.
  - Se requiere hacer coincidir al líder con la situación, o cambiar la situación para hacerla favorable al estilo del líder.
  
  El estilo del líder es su comportamiento habitual, se evalúa mediante un cuestionario, puede ser orientado a la tarea o a las relaciones.
  La situación se evalúa mediante 3 dimensiones:
  - Relaciones líder-miembro: grado de intimidad y confianza entre líder y subordinados.
  - Estructura del puesto: grado en que se definen los procedimientos para los deberes del puesto.
  - Posición de poder: grado en que el líder tiene autoridad formal para recompensar o castigar a los subordinados.

    Entre mejor sean las relaciones líder-miebro, más estructurado el puesto y mayor el poder del líder, más favorable es la situación.
    \* Se asume que el estilo de liderazgo es fijo, si el líder y la situación no coinciden, se debe cambiar al líder o trasladarlo a otro lugar.

- Teoría de liderazgo situacional: se postula que las situaciones se dan por las capacidades y disposición de los subordinados, el comportamiento más eficaz del líder depende de la motivación de los subordinados.
    Se distinguen 4 estilos de liderazgo:
    - Dirigir: alto en tareas y bajo en relaciones, subordinados poco capaces y poco dispuestos. El líder define roles, encargados y tareas.
    - Persuadir: alto en tareas y relaciones, subordinados incapaces pero dispuestos. El líder expone un comportamiento directivo y solidario.
    - Participar: bajo en tareas y alto en relaciones, subordinados capaces pero poco dispuestos. El líder y subordinados toman desiciones conjuntas. El líder facilita y comunica.
    - Delegar: bajo en tareas y relaciones, subordinados capaces y dispuestos. El líder proporciona poca dirección o apoyo.

- Teoría del camino-meta: sostiene que el trabajo del líder es ayudar a sus seguidores a alcanzar sus metas, o el apoyo necesario para asegurar que dichas metas sean compatibles con aquellas de la organización. Los líderes asumen distintos estilos en diferentes momentos dependiendo de la situación:
  - Directivo: se enfoca en el trabajo que hay que hacer.
  - Solidario: foco en las necesidades de los subordinados.
  - Participativo: consulta antes de tomar desiciones.
  - Orientado a logros: establece metas desafiantes y espera que los subordinados las cumplan.
    Hay 2 fuentes de contingencia:
    - Ambiental: estructura de las tareas, sistema formal de autoridad, grupo de trabajo.
    - Subordinada: locus de control, experiencia, habilidad percibida.
    \* No hay una mejor manera de liderar, depende de la situación. El estilo más efectivo es contingente o dependiente de la situación.
    \* Los buenos líderes modifican su comportamiento según la situación.


## Enfoques modernos
Hoy se hace la diferencia entre líder y administrador.
El gerente es el que planifica, organiza, coordina y controla. En la realidad el trabajo es más complejo, deben ser capaces de jugar varios tipos de roles.
El líder es uno de los varios roles del gerente.

Estilos de liderazgo:
- Transformacional: inspira a los seguidores a comprometerse con una visión compartida que proporciona, visionarios y carismático, acepta con entusiasmo la complejidad y ambigüedad (líder tradicional).
- Transaccional: provee estructura, planifica y controla, exhibe preocupación y provee soporte interpersonal (administrador).

_\* El liderazgo no ocurre en aislamiento._ Es igual de importante el líder al primer seguidor y la multitud.

##### Teoría de atribución
Afirma que lo que se considera como liderazgo es más bien la naturaleza perceptual (está en los ojos del espectador).

##### Líderes servidores
Plantea la motivación principal en el deseo de servir, sin intereses propios, se enfoca en las necesidades de otros.

Tiene 9 características principales:
- Empatía.
- Crear valores para la comunidad.
- Habilidades conceptuales.
- Empoderamiento.
- Ayudar a los subordinados a desarrollarse y ser exitosos.
- Poner a los subordinados primero.
- Comportarse éticamente.
- Orientado a las relaciones.
- Servidumbre.

Los estudios muestran que el liderazgo servil presenta mejoras en:
- Actitudes.
- Desempeño.
- Poder de equipo.
- Servicio al cliente.
- Identificación con la unidad organizacional.

\* Hoy los jóvenes valoran el liderazgo servil, incluso tienen la tendencia a cambiarse de trabajo de lugares con un liderazgo más autoritario y tradicional (top-down).

# Comportamiento en grupos

Se definen 2 tipos de grupos:
- Formales: definidos por la estructura de la organización, tienen tareas asignadas, se reporta al gerente asignado.
- Informales: se crean espontáneamente de acuerdo a la necesidad social del momento. Los miembros trabajan juntos para lograr un objetivo común. Es más amistoso, reúne a gente que tienen características similares.

### Group development

##### Modelo tradicional de 5 etapas
1. Formación: los miembros sienten mucha incerteza, las interacciones tienden a ser cortas y cautelosas. Hay muchas exploración sobre la tarea.
2. Tormenta: mucho desacuerdo y conflicto entre los miembros del equipo. Las interacciones son marcadas por el criticismo, respuestas emocionales, resistencia y polarización.
3. Normalización: los miembros desarrollan relaciones más estrechas y cohesivas.
4. Desempeño: el grupo se vuelve completamente funcional hacia lograr la tarea, cooperación mutua.
5. Terminación: en grupos temporales, se caracteriza por la preocupación por finalizar las actividades en vez de enfocarse en el desempeño.

\* Los equipos pueden no progresar secuencialmente entre las 5 etapas, y la etapa de desempeño no siempre es la más fructífera.

##### Punctuated equilibrium
Aplicable para grupos temporales con un plazo límite.

- En la primera reunión se define la dirección del grupo.
- Al medio tiempo ocurren cambios importantes y aumenta el desempeño drásticamente.
- Existe otro aumento de desempeño al acercarse la fecha límite.

### Elementos fundamentales de los grupos
- Roles: conjunto de patrones de comportamiento esperado.
- Normas: estándares aceptables de comportamiento dentro de un grupo y que son compartidos por sus miembros.
- Status: posición definida socialmente o rango que dan los demás a los grupos o sus miembros.
- Tamaño: número de integrantes.
- Cohesión: grado en que los miembros se ven atraídos entre sí y están motivados para permanecer en el grupo.

\* Las normas inducen conformidad, existe un deseo de aceptación, los resultados varían por cultura.

\* Hay 2 tipos de conformidad: nivel cognitivo (realmente creo que el grupo tiene razón), y nivel social (no creo lo que dice el grupo, pero sigo la corriente para no ser rechazado).

\* Ir en contra de las normas es caro, requiere un esfuerzo y un riesgo social.


##### Pereza social
Los grupos grander tienen a inducir _social loafing_ (tendencia de los individuos a reducir su esfuerzo cuando están en grupo vs. cuando trabajan individualmente).

\* Retornos decrecientes al aumentar el tamaño del grupo.

\* El desempeño grupal **no** es la suma de los desempeños individuales.

### Toma de desiciones en grupo.

Fortalezas:
- Genera información y conocimiento más completos.
- Ofrece mayor diversidad de perspectivas.
- Aumenta la aceptación de las desiciones.

Debilidades:
- Consume más tiempo.
- Hay presión por conformidad (potencial _groupthinking_).
- Las discusiones pueden ser dominadas por unos pocos.
- Puede haber responsabilidad ambigua. 

##### Groupthinking
Situaciones en las que presiones por conformidad impiden que el grupo evalúe objetivamente las visiones inusuales, minoritarias, impopulares o disidentes.

Síntomas:
- Ilusión de invulnerabilidad.
- Racionalización colectiva.
- Creencia de la inherente moralidad del grupo interno.
- Estereotipado de los grupos externos.
- Presiones sobre quienes disienten.
- **Auto-censura**: los miembros con dudas o puntos de vista distintos guardan silencio para evitar desviarse del consenso del grupo.
- **Ilusión de alta cohesión y unanimidad**: los miembros asumen que todos están de acuerdo con la desición del grupo, cuando en realidad muchos tienen dudas o reservas.

Prevención:
- Líderes deben evitar expresar sus opiniones al inicio.
- Designar a un "abogado del diablo".
- Dividir el grupo en subgrupos independientes.
- Buscar opiniones externas.
- Realizar sesiones de retroalimentación después de la desición.



# cultura

La cultura es el conjunto de valores, creencias, entendimientos, y formas de pensar que son compartidas por los miembros de una organización, y que es enseñada a los nuevos miembros como lo que es correcto.

Provee a los individuos de un sentido de identidad organizacional, genera compromiso con la organización. Ayuda a los nuevos integrantes a hacer sentido de lo que es y lo que hace la organización.

Informa _la forma en que hacemos las cosas aquí_.

Hay 2 niveles:
- Observable: símbolos, ceremonias, historias, slogans, comportamientos, vestimenta, disposición física.
- Subyacente: valores, supuestos, creencias, actitudes, sentimientos.

### Observables
Representan valores subyacentes profundos que residen en la mente de los miembros de la organización.

Algunos observables se manifiestan en patrones de interacción social que se dan en las actividades que realiza la organización.

Para identificar e interpretar la cultura organizacional se requiere hacer inferencias basándose en observables y patrones de interacción social.

##### Ritos y ceremonias
Son ocasiones especiales para reforzar valores específicos, crear entendimientos compartidos, y ungir y celebrar _héroes_ que **simbolicen** valores importantes.
Los gerentes pueden usarlas para proveer ejemplos patentes de qué es lo que la empresa valora.

##### Historias
Narrativas basadas en eventos reales, frecuentemente usadas para informar a los nuevos miembros acerca de la organización y para mantener vivos sus valores primarios. Incluye mitos y leyendas de los héroes de la organización.

##### Símbolos
Puede ser una ceremonia, una historia, o un artefacto físico que simbolice/represente valores profundo de la organización.

##### Lenguaje
Slogans, dichos, metáforas, etc. para transmitir ciertos significados especiales a los empleados.

### Valores subyacentes
Son los valores centrales que son compartidos por los miembros de la organización, y que son difíciles de cambiar.
Son los que realmente conforman la cultura organizacional.

### Intensidad de la cultura
- Intensidad: denota el alineamiento entre los miembros de la organización acerca de la importancia de ciertos valores específicos.

- Culturas fuentes:
  - Existe un consenso acerca de la importancia de los valores
  - Los valores fundamentales están profundamente arraigados y son muy compartidos.
  - Uso frecuente de símbolos, historias, héroes y lenguaje propio.
  - La cultura es cohesiva e intensa y ejerce una mayor influencia en los miembros de la organización.

- Culturas débiles: 
  - La cultura organizacional no afecta mucho el comportamientos de los empleados.
  - No hay mucho compromiso ni aceptación por parte de los empleados.
  - La organización no hace mucho esfuerzo por fortalecerla.

Algunos factores que influyen en la intensidad de la cultura:
- Índice de rotación de los empleados: a mayor rotación, menor intensidad.
- Fortaleza de la cultura original.
- Claridad de los valores y creencias estructurales.
- Uso de rituales, símbolos, difusión de mitos, lenguaje, etc.

### Cultura como un arma estratégica
Además de usar símbolos, historais, lenguaje y ceremonias, los gerentes se apoyan en la selección y socialización de nuevos empleados (seleccionar solo empleados que tienen el potencial de alinearse con la cultura, y usar los artefactos durante el período de inducción).

La cultura correcta puede hacer que se necesite menos supervisión y administración (cultura de ética, cultura competitiva, los empleados se autogestionan y/o supervisan a los demás), ojo de no caer al extremo (iron tightening cage).

### Fuentes de la cultura
- Los fundadores
- Visión y misión
- Prácticas pasadas de la organización
- El comportamiento de la alta administración
- Perpetuación de la cultura organizacional

### Tipos de cultura
Hay 2 dimensiones básicas:
- Foco interno e integración vs. foco externo y diferenciación.
- Estabilidad y control vs. flexibilidad y discreción.

##### Clan
Foco interno y flexibilidad.

- Ambiente amigable y familiar
- Lealtad y tradición esenciales.
- Los líderes son mentores y protectores.
- Gran valor a la cohesión y clima laboral.
- Énfasis al desarrollo personal a largo plazo.
- Éxito definido por la sensibilidad frente a los clientes y preocupación por las personas.
- Se reconoce el trabajo en equipo, la participación y el consenso.

##### Jerárquica (burocrática)
Foco interno y control.

- Lugar estructurado y estandarizado para trabajar.
- Los líderes son eficientes coordinadores y organizadores.
- Los procedimientos cobiernan lo que las personas hacen.
- Es crítico mantener la organización funcionando como reloj.
- La preocupación es sobre estabilidad y operación eficiente.
- Gran valor a las reglas y políticas.
- Éxito se mide por la confiabilidad en las entregas, fluida programación y bajos costos.

##### Mercado
Foco externo y control.

- Organización orientada a resultados, cuya mayor perocupación es realizar el trabajo.
- Las personas son competitivas y orientadas a metas.
- Los líderes son conductores duros, productivos y exigentes.
- Énfasis en ganar.
- La reputación y el éxito son preocupaciones corrientes.
- La preocupación a largo plazo es desarrollar acciones competitivas para alcanzar logros medibles.
- Éxito se mide en participación y penetración de mercado.
- Precios y liderazgos de mercado son importantes.
- Estilo organizacional es una conducción de alta competitividad.

##### Adhocrática (adaptable)
Foco externo y flexibilidad.

- Lugar dinámico, emprendedor y creativo para trabajar.
- Personas corren riesgos.
- Líderes considerados innovadores y tomadores de riesgo.
- Se valora el compromiso, experimentación e innovación.
- Énfasis en liderazgo de innovación.
- Énfasis a largo plazo en crecimiento y adquirir nuevos recursoso.
- Éxito se mide en productos y servicios únicos.
- Importante ser un líder en servicios y productos.
- Se reconoce la iniciativa y autonomía individual.

# Transformación digital

Se consideran 4 revoluciones industriales:
1. Carbón, máquina a vapor, mecanización, fábricas, trenes.
2. Electricidad, petróleo, telégrafo, líneas de producción, aviones.
3. Electrónica, computación, digitalización.
4. (Actual) Internet, IA, big data.

La transformación digital involucra capacidades tecnológicas y aspectos administrativos:
- Rediseño de estructura organizacional.
- Cambio de cultura organizacional.
- Alineamiento con la estrategia.
- Cambios en modelos de negocios (en los casos más extremos).
- Cambios regulatorios (a nivel gubernamental).

También hay diversos grados de transformación, desde proyectos únicos (one-offs), hasta organizaciones disruptivas (singularity productivity) que hacen transformación constante y gobierno digital.

Hay distintas tecnologías de gestión de la innovación (Agile, Lean, Design thinking).

### AI
- Strong AI: máquinas que piensan y actúan de forma que igualan o superan a los humanos. Hasta ahora es ficción.
- Weak AI: cualquier actividad que pueda hacer un computador que haya sido hecha alguna vez por un humano. Es lo que las organizaciones están implementando actualmente.

### Transformaciones, progreso y disrupción
En cada revolución tecnológica ha habido resistencia al cambio, y un pensamiento de _la tecnología nos reemplazará._ La verdad es que la economía se reconvierte, surgen ocupaciones y servicios que antes no existían.

Si habrá disrupción, las organizaciones que no logren adecuarse darán paso a aquellos que sí (selección natural organizacional).

### Bounded automation
- Pricing of labour: los trabajadores con menor sueldo son menos probable de ser reemplazador por automatización. Mientras más alto el pago, más probable es que atraiga automatización.

- Professional power: el poder de una profesión va a determinar cómo la tecnología se diseña y usa (un médico tiene más poder de determinar cómo se diseña y usa la tecnología, vs. un camionero).

- Nature of task: mientras más rutinaria la tarea, más probable es que sea automatizada.
